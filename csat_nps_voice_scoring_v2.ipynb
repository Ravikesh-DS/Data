|---|---|

| **B. Text sentiment component** | Text score per interaction (plug-in from Sentiment Agent, or a baseline model trained on a customer-grouped split) |

````python
for d in (SILVER_DIR, GOLD_DIR, OUT_DIR):
    d.mkdir(parents=True, exist_ok=True)
try:
    con.close()
except NameError:
    pass
con = duckdb.connect(str(WORK_DB))
con.execute(f"SET threads={DUCKDB_THREADS}")
con.execute(f"SET memory_limit='{DUCKDB_MEMORY}'")
con.execute(f"SET temp_directory='{(OUT_DIR / 'duckdb_tmp').as_posix()}'")
con.execute("SET preserve_insertion_order=false")

FINDINGS, RESULTS = [], {}


def q(sql):
    return con.sql(sql).df()


def sql_list(v):
    return "(" + ", ".join(f"'{x}'" for x in v) + ")"


def save(df, name, show=SHOW_ROWS):
    df.to_csv(OUT_DIR / f"{name}.csv", index=False)
    print(f"--- {name} ({len(df):,} rows) -> {OUT_DIR / (name + '.csv')}")
    if show:
        display(df.head(show))
    return df


def flag(tag, msg):
    FINDINGS.append((tag, msg))
    print(f"  [FLAG] {msg}")


def table_exists(name):
    return bool(q(f"SELECT count(*) FROM information_schema.tables WHERE table_name='{name}'").iloc[0, 0])


def cramers_v(ct):
    chi2, p, _, _ = chi2_contingency(ct)
    n = ct.values.sum()
    k = min(ct.shape) - 1
    return float(np.sqrt(chi2 / (n * k))) if k > 0 else 0.0, float(p)


print("DuckDB", duckdb.__version__, "| pandas", pd.__version__, "| outputs ->", OUT_DIR.resolve())
````

````python
def src(t):
    f, d = BRONZE_DIR / f"{t}.parquet", BRONZE_DIR / t
    if f.exists():
        return f"read_parquet('{f.as_posix()}')"
    if d.exists():
        return f"read_parquet('{(d / '**' / '*.parquet').as_posix()}')"
    return None

for t in ("customers", "interactions", "engagements", "transactions"):
    s = src(t)
    if s is None and t in ("customers", "interactions"):
        raise FileNotFoundError(f"{t} parquet not found in {BRONZE_DIR.resolve()}")
    if s:
        con.execute(f"CREATE OR REPLACE VIEW raw_{t} AS SELECT * FROM {s}")
HAS_TX, HAS_ENG = src("transactions") is not None, src("engagements") is not None
print("transactions:", HAS_TX, "| engagements:", HAS_ENG)

con.execute("""
CREATE OR REPLACE TABLE cust1 AS
SELECT customer_id, TRY_CAST(signup_date AS DATE) signup, loyalty_tier, customer_segment, region, preferred_channel,
       (upper(trim(CAST(marketing_consent AS VARCHAR))) IN ('Y','YES','TRUE','1')) marketing_consent
FROM raw_customers
QUALIFY row_number() OVER (PARTITION BY customer_id ORDER BY signup_date) = 1
""")

con.execute(f"""
CREATE OR REPLACE VIEW itx AS
SELECT interaction_id, customer_id, ts, CAST(date_trunc('month', ts) AS DATE) mth,
       channel, interaction_type, resolution_status,
       (upper(trim(CAST(first_contact_resolution AS VARCHAR))) IN ('Y','YES','TRUE','1')) fcr,
       csat_raw, nps_raw,
       CASE WHEN csat_raw BETWEEN 1 AND 5 AND csat_raw = round(csat_raw) THEN csat_raw END csat,   -- invalid -> NULL (not repaired)
       CASE WHEN nps_raw BETWEEN 0 AND 10 AND nps_raw = round(nps_raw) THEN nps_raw END nps,
       CAST(interaction_text AS VARCHAR) txt, sentiment_label, TRY_CAST(sentiment_score_gt AS DOUBLE) sscore
FROM (SELECT *, TRY_CAST(interaction_ts AS TIMESTAMP) ts,
             TRY_CAST(csat_score AS DOUBLE) csat_raw, TRY_CAST(nps_score AS DOUBLE) nps_raw
      FROM raw_interactions)
WHERE ts >= TIMESTAMP '{WINDOW_START}' AND ts < TIMESTAMP '{WINDOW_END}'
""")

tx_sql = (f"""SELECT customer_id, CAST(date_trunc('month', TRY_CAST(transaction_ts AS TIMESTAMP)) AS DATE) mth, count(*) n_tx
              FROM raw_transactions GROUP BY ALL""" if HAS_TX else
          "SELECT CAST(NULL AS VARCHAR) customer_id, CAST(NULL AS DATE) mth, 0 n_tx WHERE false")
eng_sql = (f"""SELECT customer_id, CAST(date_trunc('month', TRY_CAST(event_ts AS TIMESTAMP)) AS DATE) mth, count(*) n_eng
               FROM raw_engagements GROUP BY ALL""" if HAS_ENG else
           "SELECT CAST(NULL AS VARCHAR) customer_id, CAST(NULL AS DATE) mth, 0 n_eng WHERE false")
con.execute(f"""
CREATE OR REPLACE TABLE act_m AS
WITH t AS ({tx_sql}), e AS ({eng_sql}),
     i AS (SELECT customer_id, mth, count(*) n_int FROM itx GROUP BY ALL),
     te AS (SELECT coalesce(t.customer_id, e.customer_id) customer_id, coalesce(t.mth, e.mth) mth,
                   coalesce(n_tx, 0) n_tx, coalesce(n_eng, 0) n_eng
            FROM t FULL JOIN e ON t.customer_id = e.customer_id AND t.mth = e.mth)
SELECT coalesce(te.customer_id, i.customer_id) customer_id, coalesce(te.mth, i.mth) mth,
       coalesce(n_tx, 0) n_tx, coalesce(n_eng, 0) n_eng, coalesce(n_int, 0) n_int
FROM te FULL JOIN i ON te.customer_id = i.customer_id AND te.mth = i.mth
WHERE coalesce(te.mth, i.mth) >= DATE '{WINDOW_START}' AND coalesce(te.mth, i.mth) < DATE '{WINDOW_END}'
""")
display(q("SELECT (SELECT count(*) FROM cust1) customers, (SELECT count(*) FROM itx) interactions, (SELECT count(*) FROM act_m) active_customer_months"))
````

### A3. Missingness mechanism – is the response rate the same everywhere?
If response rates do not depend on channel, type, sentiment, resolution, month or tier, responses are **missing (close to) completely at random (MCAR)** on observables → survey averages are unbiased and *presence of a survey carries no information* (never use “has NPS” as a feature).
With 1M rows tiny differences are “significant”, so use **effect size (Cramér’s V; <0.02 negligible)** and the **spread in percentage points**.

````python
DIMS = ["channel", "interaction_type", "resolution_status", "sentiment_label", "fcr", "mth",
        "loyalty_tier", "customer_segment", "region"]
con.execute("CREATE OR REPLACE TEMP VIEW itx_c AS SELECT i.*, c.loyalty_tier, c.customer_segment, c.region FROM itx i LEFT JOIN cust1 c USING (customer_id)")
rows, frames = [], []
for d in DIMS:
    t = q(f"""SELECT CAST({d} AS VARCHAR) level, count(*) n, count(nps) n_nps, count(csat) n_csat,
                     round(100.0*count(nps)/count(*), 2) nps_resp_pct, round(100.0*count(csat)/count(*), 2) csat_resp_pct,
                     round(avg(nps), 3) avg_nps, round(avg(csat), 3) avg_csat,
                     CASE WHEN count(nps) >= {MIN_N_INDEX}
                          THEN round(100.0*(count_if(nps >= 9) - count_if(nps <= 6))/count(nps), 1) END nps_index
              FROM itx_c GROUP BY 1 ORDER BY 1""")
    frames.append(t.assign(dimension=d))
    out = {"dimension": d, "levels": len(t)}
    for s, col in (("nps", "n_nps"), ("csat", "n_csat")):
        ct = pd.DataFrame({"yes": t[col], "no": t.n - t[col]})
        v, p = cramers_v(ct)
        out.update({f"{s}_spread_pp": round(t[f"{s}_resp_pct"].max() - t[f"{s}_resp_pct"].min(), 2),
                    f"{s}_cramers_v": round(v, 4), f"{s}_p_value": p})
    rows.append(out)
save(pd.concat(frames)[["dimension", "level", "n", "n_nps", "nps_resp_pct", "n_csat", "csat_resp_pct",
                        "avg_nps", "avg_csat", "nps_index"]], "A3_response_rate_by_dimension", show=0)
MCAR = save(pd.DataFrame(rows), "A3_missingness_summary", show=20)
RESULTS["max_nps_spread_pp"] = MCAR.nps_spread_pp.max()
bad = MCAR[(MCAR.nps_spread_pp > MCAR_SPREAD_PP) | (MCAR.nps_cramers_v >= 0.02)]
if len(bad):
    flag("A3", "NPS response rate varies with: " + ", ".join(bad.dimension) + " -> responses NOT random; weight/report by group")
else:
    flag("A3", f"NPS response rate flat across all {len(DIMS)} dimensions (max spread {MCAR.nps_spread_pp.max()} pp, "
               f"max V {MCAR.nps_cramers_v.max()}) -> MCAR on observables; survey presence is uninformative")
````

### A6. What drives NPS / CSAT? (interaction vs relationship)
Variance explained (R²) by interaction attributes and by the ground-truth text sentiment. High R² ⇒ NPS here is largely a **transactional (interaction-level) score**.

````python
for d in ["interaction_type", "resolution_status", "fcr", "sentiment_label"]:
    save(q(f"""SELECT CAST({d} AS VARCHAR) {d}, count(nps) n_nps, round(avg(nps), 2) avg_nps, round(avg(csat), 2) avg_csat,
                      round(100.0*count_if(nps >= 9)/nullif(count(nps), 0), 1) promoter_pct,
                      round(100.0*count_if(nps <= 6)/nullif(count(nps), 0), 1) detractor_pct,
                      CASE WHEN count(nps) >= {MIN_N_INDEX}
                           THEN round(100.0*(count_if(nps >= 9) - count_if(nps <= 6))/count(nps), 1) END nps_index
               FROM itx GROUP BY 1 ORDER BY avg_nps"""), f"A6_nps_by_{d}")

r6 = q("SELECT interaction_type, resolution_status, fcr, sentiment_label, sscore, nps, csat FROM itx WHERE nps IS NOT NULL")
var = r6.nps.var(ddof=0)
r2 = lambda g: 1 - ((r6.nps - r6.groupby(g).nps.transform("mean")) ** 2).mean() / var
rows = [{"explained_by": " + ".join(g), "r2": round(r2(g), 3)} for g in
        (["interaction_type"], ["resolution_status"], ["fcr"], ["sentiment_label"], ["interaction_type", "sentiment_label"],
         ["interaction_type", "sentiment_label", "resolution_status"])]
b = np.polyfit(r6.sscore, r6.nps, 1)
rows.append({"explained_by": "sscore (linear)", "r2": round(1 - (r6.nps - np.polyval(b, r6.sscore)).var(ddof=0) / var, 3)})
R2 = save(pd.DataFrame(rows), "A6_nps_variance_explained")
both = q("SELECT csat, nps, sscore FROM itx WHERE csat IS NOT NULL AND nps IS NOT NULL")
save(both.corr(method="spearman").round(3).reset_index(), "A6_spearman_csat_nps_sentiment")
RESULTS["corr_csat_nps"] = both.corr(method="spearman").loc["csat", "nps"]
flag("A6", f"Text sentiment explains {R2.r2.iloc[-1]:.0%} of NPS variance; Spearman(CSAT,NPS)={RESULTS['corr_csat_nps']:.2f} "
           "-> NPS is interaction-driven; CSAT and NPS overlap (cap combined survey weight)")
````

# Part B – Text sentiment component (input to the Sentiment Score)
* **Customer-grouped split** (`hash(customer_id) % 100 < TEST_PCT` → test) is assigned once and stored in the data model, so every downstream metric is out-of-sample by customer.
* If `TEXT_PRED_PATH` exists (Sentiment Agent / LLM output: `interaction_id, text_sentiment_pred`), it is used as-is.
* Otherwise a **baseline** (TF-IDF + Ridge on `sentiment_score_gt`, trained on **train customers only**) is fitted. The GT columns are labels – they never enter the score directly.
* Texts are highly templated, so a second **template hold-out** (unseen texts) is also evaluated in Part E.

````python
con.execute(f"""CREATE OR REPLACE TABLE split_c AS
                SELECT customer_id, CASE WHEN hash(customer_id) % 100 < {TEST_PCT} THEN 'test' ELSE 'train' END split FROM cust1""")
display(q("SELECT split, count(*) customers FROM split_c GROUP BY 1"))

n_texts = q("SELECT count(DISTINCT txt) FROM itx").iloc[0, 0]
print(f"distinct texts: {n_texts:,}  (interactions: {q('SELECT count(*) FROM itx').iloc[0,0]:,})")
if n_texts < 5000:
    flag("B", f"Only {n_texts:,} distinct texts -> templated text; text metrics on random splits are optimistic (see V1 template hold-out)")

LABELS = ["negative", "neutral", "positive"]

def fit_text_model(train_df):
    """train_df: txt, n, y (mean GT score), lab. Returns (vectorizer, model, (lo, hi) label cut-offs)."""
    vec = TfidfVectorizer(ngram_range=(1, 2), sublinear_tf=True, min_df=1)
    X = vec.fit_transform(train_df.txt)
    mdl = Ridge(alpha=1.0).fit(X, train_df.y, sample_weight=train_df.n)
    pred = np.clip(mdl.predict(X), -1, 1)
    best = (-1, (-0.2, 0.2))
    for lo in np.arange(-0.6, 0.001, 0.02):           # cut-offs tuned on TRAIN only
        for hi in np.arange(0.0, 0.601, 0.02):
            lab = np.where(pred < lo, "negative", np.where(pred > hi, "positive", "neutral"))
            f = f1_score(train_df.lab, lab, average="macro", sample_weight=train_df.n)
            if f > best[0]:
                best = (f, (round(lo, 2), round(hi, 2)))
    return vec, mdl, best[1]

def to_label(pred, cuts):
    lo, hi = cuts
    return np.where(pred < lo, "negative", np.where(pred > hi, "positive", "neutral"))

texts = q("SELECT DISTINCT txt FROM itx WHERE txt IS NOT NULL")
if TEXT_PRED_PATH.exists():
    TEXT_SOURCE = "plug-in"
    con.execute(f"""CREATE OR REPLACE TABLE text_pred_i AS
                    SELECT interaction_id, greatest(-1, least(1, CAST(text_sentiment_pred AS DOUBLE))) text_pred
                    FROM read_parquet('{TEXT_PRED_PATH.as_posix()}')""")
    tr = q("""SELECT p.text_pred, i.sentiment_label lab FROM text_pred_i p JOIN itx i USING (interaction_id)
              JOIN split_c s USING (customer_id) WHERE s.split = 'train'""")
    best = (-1, None)
    for lo in np.arange(-0.6, 0.001, 0.02):
        for hi in np.arange(0.0, 0.601, 0.02):
            f = f1_score(tr.lab, to_label(tr.text_pred.values, (lo, hi)), average="macro")
            if f > best[0]:
                best = (f, (round(lo, 2), round(hi, 2)))
    CUTS = best[1]
else:
    TEXT_SOURCE = "baseline TF-IDF+Ridge"
    train_t = q("""SELECT i.txt, count(*) n, avg(i.sscore) y, mode(i.sentiment_label) lab
                   FROM itx i JOIN split_c s USING (customer_id) WHERE s.split = 'train' AND i.txt IS NOT NULL GROUP BY 1""")
    VEC, MDL, CUTS = fit_text_model(train_t)
    texts["text_pred"] = np.clip(MDL.predict(VEC.transform(texts.txt)), -1, 1)
    texts["text_label_pred"] = to_label(texts.text_pred.values, CUTS)
    con.register("texts_df", texts)
    con.execute("CREATE OR REPLACE TABLE text_lookup AS SELECT * FROM texts_df")
    con.unregister("texts_df")

print("text source:", TEXT_SOURCE, "| label cut-offs (tuned on train):", CUTS)
````

# Part C – Non-imputed scoring specification
**Normalisation to −1…+1** (observed values only):

$$\text{CSAT}_{norm}=\frac{\text{CSAT}-3}{2}\qquad \text{NPS}_{norm}=\frac{\text{NPS}-5}{5}$$

**Observed-only weights for customer *c*, month *m*:** base weights $w_{text}=0.6,\ w_{csat}=0.2,\ w_{nps}=0.2$; a component gets weight only if it was observed that month.
If CSAT and NPS are both observed, their combined weight is capped at `SURVEY_CAP` (0.30) to avoid double counting.

$$S_{cm}=\frac{\sum_{k\in \text{observed}} w_k\,\bar{x}_k}{\sum_{k\in \text{observed}} w_k},\qquad k\in\{\text{text},\ \text{CSAT},\ \text{NPS}\}$$

| Situation | Effective weights | Score |
|---|---|---|
| Text only | text 1.00 | text average |
| Text + CSAT | text 0.75, CSAT 0.25 | weighted |
| Text + NPS | text 0.75, NPS 0.25 | weighted |
| Text + CSAT + NPS (cap 0.30) | text 0.667, CSAT 0.167, NPS 0.167 | weighted |
| Nothing observed | – | **NULL – `not_available` (never neutral 0)** |

**Not allowed:** filling missing CSAT/NPS with 0/mean/median/model, carrying last NPS forward into the score, using “has survey” as a feature, using tier/segment in the score.
**Allowed:** observed counts, three-state look-back context flags (`not_surveyed` is its own state), last observed NPS shown as *labelled context*.

**Conflict:** text and survey have opposite signs and differ by ≥ `CONFLICT_GAP` → flagged, confidence downgraded one level.

**Confidence band** – `high`: text + ≥1 survey, no conflict · `medium`: ≥2 interactions text-only, or text+survey with conflict, or survey without text · `low`: 1 interaction text-only · `not_available`: nothing observed.
**Confidence score (0–1)** = 0.5 × coverage (observed base weight ÷ total) + 0.3 × volume (min(n_text,3)/3) + 0.2 × agreement (1 − |text − survey|/2; 0.5 if no survey to verify).

**Voice actions (NPS triggers use the month’s *minimum* NPS):** cancellation + negative → `retention_save_desk`; detractor or CSAT ≤ 2 → `service_recovery_48h`; score < −0.3 (medium/high conf.) → `proactive_service_check`; promoter & score ≥ 0.5 → `advocacy_referral_invite` (needs marketing consent, else `thank_you_service_only`); passive → `nurture_offer` (needs consent); else `monitor`.

### C1. Python reference implementation (used for worked examples, unit tests and the weight grid in Part E)

````python
def composite(text_avg, n_text, csat_norm, n_csat, nps_norm, n_nps, w=None, cap=SURVEY_CAP):
    """Vectorised observed-only composite. Returns DataFrame: score, survey_score, w_text_eff, w_csat_eff, w_nps_eff."""
    w = w or WEIGHTS
    n_text, n_csat, n_nps = (np.asarray(x, dtype=float) for x in (n_text, n_csat, n_nps))
    t, c, n = (np.asarray(x, dtype=float) for x in (text_avg, csat_norm, nps_norm))
    wt = np.where(n_text > 0, w["text"], 0.0)
    wc = np.where(n_csat > 0, w["csat"], 0.0)
    wn = np.where(n_nps > 0, w["nps"], 0.0)
    cap = np.inf if cap is None else cap
    scale = np.where(wc + wn > cap, cap / np.where(wc + wn > 0, wc + wn, 1), 1.0)
    wc, wn = wc * scale, wn * scale
    wsum = wt + wc + wn
    num = np.where(wt > 0, wt * np.nan_to_num(t), 0) + np.where(wc > 0, wc * np.nan_to_num(c), 0) \
        + np.where(wn > 0, wn * np.nan_to_num(n), 0)        # nan_to_num only where weight is 0 -> term excluded, not imputed
    with np.errstate(invalid="ignore", divide="ignore"):
        score = np.where(wsum > 0, num / wsum, np.nan)
        ssum = wc + wn
        survey = np.where(ssum > 0, (np.where(wc > 0, wc * np.nan_to_num(c), 0) + np.where(wn > 0, wn * np.nan_to_num(n), 0)) / ssum, np.nan)
        return pd.DataFrame({"score": score, "survey_score": survey,
                             "w_text_eff": np.where(wsum > 0, wt / wsum, np.nan),
                             "w_csat_eff": np.where(wsum > 0, wc / wsum, np.nan),
                             "w_nps_eff": np.where(wsum > 0, wn / wsum, np.nan)})


def confidence(n_text, text_avg, survey_score, n_csat, n_nps, w=None):
    w = w or WEIGHTS
    n_text, n_csat, n_nps = (np.asarray(x, dtype=float) for x in (n_text, n_csat, n_nps))
    t, s = np.asarray(text_avg, dtype=float), np.asarray(survey_score, dtype=float)
    has_t, has_s = n_text > 0, ~np.isnan(s)
    conflict = has_t & has_s & (np.nan_to_num(t) * np.nan_to_num(s) < 0) & (np.abs(np.nan_to_num(t) - np.nan_to_num(s)) >= CONFLICT_GAP)
    band = np.select([~has_t & ~has_s, has_t & has_s & ~conflict, has_t & has_s & conflict, ~has_t & has_s, n_text >= 2],
                     ["not_available", "high", "medium", "medium", "medium"], default="low")
    cov = (np.where(has_t, w["text"], 0) + np.where(n_csat > 0, w["csat"], 0) + np.where(n_nps > 0, w["nps"], 0)) / sum(w.values())
    agree = np.where(has_t & has_s, 1 - np.abs(np.nan_to_num(t) - np.nan_to_num(s)) / 2, 0.5)
    score = np.where(band == "not_available", np.nan, 0.5 * cov + 0.3 * np.minimum(n_text, 3) / 3 + 0.2 * agree)
    return pd.DataFrame({"confidence_band": band, "confidence_score": np.round(score, 3), "conflict": conflict})


ex = pd.DataFrame([
    ("1 interaction, text only",            1, 0.40, np.nan, 0, np.nan, 0),
    ("3 interactions, text only",           3, -0.30, np.nan, 0, np.nan, 0),
    ("text + CSAT 2",                        1, -0.20, -0.5, 1, np.nan, 0),
    ("text + NPS 3 (detractor)",             2, 0.10, np.nan, 0, -0.4, 1),
    ("text + CSAT 5 + NPS 10 (cap applies)", 1, 0.60, 1.0, 1, 1.0, 1),
    ("conflict: positive text, NPS 1",       1, 0.65, np.nan, 0, -0.8, 1),
    ("no interaction, no survey",            0, np.nan, np.nan, 0, np.nan, 0),
], columns=["case", "n_text", "text_avg", "csat_norm", "n_csat", "nps_norm", "n_nps"])
cmp_ = composite(ex.text_avg, ex.n_text, ex.csat_norm, ex.n_csat, ex.nps_norm, ex.n_nps)
cf_ = confidence(ex.n_text, ex.text_avg, cmp_.survey_score, ex.n_csat, ex.n_nps)
display(pd.concat([ex, cmp_.round(3), cf_], axis=1))
````

# Part D – Customer-month data model
```
dim_customer (1 row / customer, CONTEXT ONLY)            fact_interaction_scored (1 row / interaction)       [silver]
  customer_id PK, signup, loyalty_tier, segment,  ──1:N──▶  interaction_id PK, customer_id, mth, split,
  region, preferred_channel, marketing_consent              type, resolution, fcr, text_pred, text_label_pred,
                                                            csat, nps, csat_norm, nps_norm, nps_category (observed only)
            │                                                          │ aggregate (no imputation)
            ▼                                                          ▼
customer_month_voice  (1 row / customer_id × mth; spine = every month from signup, Oct-25…Sep-26)       [gold]
  keys · context(only) · activity coverage · interaction evidence · survey evidence (NULL when not observed)
  · look-back context (3-state) · score components & effective weights · sentiment_score · confidence · drivers · voice action · lineage
            │
            ▼
agg_segment_nps_month (mth × dimension × level; NPS index suppressed when n < MIN_N_INDEX)                [gold]
```
**Design rules:** grain `customer_id × mth`; spine keeps *every* eligible month so coverage denominators are honest;
survey value columns are NULL when not observed (counts are 0); context columns never feed the score;
the next Behavioral Health table joins on the same key.

### D2. Build `customer_month_voice` (gold) – scoring, confidence, drivers and actions in SQL

````python
WT, WC, WN = WEIGHTS["text"], WEIGHTS["csat"], WEIGHTS["nps"]
WTOT = WT + WC + WN
COUNT_COLS = ["n_text", "n_text_negative", "n_issue_contacts", "n_cancel_req", "n_open_issue", "n_csat", "n_csat_low",
              "n_nps", "n_promoter", "n_passive", "n_detractor", "n_both_same_interaction", "n_nps_on_open_issue"]

SCORE_SQL = f"""
    CASE WHEN w_sum > 0 THEN (CASE WHEN n_text > 0 THEN w_t * text_avg ELSE 0 END
                            + CASE WHEN n_csat > 0 THEN w_c * csat_norm_avg ELSE 0 END
                            + CASE WHEN n_nps  > 0 THEN w_n * nps_norm_avg  ELSE 0 END) / w_sum END"""

t0 = time.time()
con.execute(f"""
CREATE OR REPLACE TABLE cm_voice AS
WITH months AS (SELECT CAST(range AS DATE) mth FROM range(DATE '{WINDOW_START}', DATE '{WINDOW_END}', INTERVAL 1 MONTH)),
spine AS (SELECT c.customer_id, m.mth FROM cust1 c CROSS JOIN months m
          WHERE c.signup IS NULL OR m.mth >= date_trunc('month', c.signup)),
iv AS (
  SELECT customer_id, mth,
         count(*) n_int, count(text_pred) n_text, avg(text_pred) text_avg, min(text_pred) text_min,
         count_if(text_label_pred = 'negative') n_text_negative,
         count_if(interaction_type IN {sql_list(ISSUE_TYPES)}) n_issue_contacts,
         count_if(interaction_type = 'cancellation_request') n_cancel_req,
         count_if(resolution_status IN {sql_list(OPEN_STATUS)}) n_open_issue,
         count(csat) n_csat, avg(csat) csat_avg, min(csat) csat_min, avg(csat_norm) csat_norm_avg, count_if(csat <= 2) n_csat_low,
         count(nps) n_nps, avg(nps) nps_avg, min(nps) nps_min, max(nps) nps_max, avg(nps_norm) nps_norm_avg,
         arg_min(nps, ts) FILTER (WHERE nps IS NOT NULL) nps_first_in_month,
         arg_max(nps, ts) FILTER (WHERE nps IS NOT NULL) nps_last_in_month,
         count_if(nps >= 9) n_promoter, count_if(nps BETWEEN 7 AND 8) n_passive, count_if(nps <= 6) n_detractor,
         count_if(csat IS NOT NULL AND nps IS NOT NULL) n_both_same_interaction,
         count_if(nps IS NOT NULL AND resolution_status IN {sql_list(OPEN_STATUS)}) n_nps_on_open_issue,
         max(ts) last_interaction_ts
  FROM fis GROUP BY ALL),
b AS (
  SELECT s.customer_id, s.mth, sp.split,
         c.loyalty_tier, c.customer_segment, c.region, c.preferred_channel, c.marketing_consent,
         coalesce(a.n_tx, 0) n_tx, coalesce(a.n_eng, 0) n_eng, coalesce(iv.n_int, 0) n_int,
         {", ".join(f"coalesce(iv.{x}, 0) {x}" for x in COUNT_COLS)},          -- counts: 0 is a fact, not an imputation
         iv.text_avg, iv.text_min, iv.csat_avg, iv.csat_min, iv.csat_norm_avg,
         iv.nps_avg, iv.nps_min, iv.nps_max, iv.nps_norm_avg, iv.nps_first_in_month, iv.nps_last_in_month,
         iv.last_interaction_ts
  FROM spine s JOIN cust1 c USING (customer_id) JOIN split_c sp USING (customer_id)
  LEFT JOIN act_m a ON a.customer_id = s.customer_id AND a.mth = s.mth
  LEFT JOIN iv ON iv.customer_id = s.customer_id AND iv.mth = s.mth),
w AS (
  SELECT *,
         CASE WHEN n_text > 0 THEN {WT} ELSE 0 END w_text_raw,
         CASE WHEN n_csat > 0 THEN {WC} ELSE 0 END w_csat_raw,
         CASE WHEN n_nps  > 0 THEN {WN} ELSE 0 END w_nps_raw
  FROM b),
w2 AS (
  SELECT *, w_text_raw w_t,
         w_csat_raw * CASE WHEN w_csat_raw + w_nps_raw > {SURVEY_CAP} THEN {SURVEY_CAP} / (w_csat_raw + w_nps_raw) ELSE 1 END w_c,
         w_nps_raw  * CASE WHEN w_csat_raw + w_nps_raw > {SURVEY_CAP} THEN {SURVEY_CAP} / (w_csat_raw + w_nps_raw) ELSE 1 END w_n
  FROM w),
sc AS (
  SELECT *, w_t + w_c + w_n w_sum FROM w2),
sc2 AS (
  SELECT *,
         {SCORE_SQL} sentiment_score,
         CASE WHEN w_c + w_n > 0 THEN (CASE WHEN n_csat > 0 THEN w_c * csat_norm_avg ELSE 0 END
                                     + CASE WHEN n_nps  > 0 THEN w_n * nps_norm_avg  ELSE 0 END) / (w_c + w_n) END survey_score
  FROM sc),
cf AS (
  SELECT *,
         coalesce(n_text > 0 AND survey_score IS NOT NULL AND text_avg * survey_score < 0
                  AND abs(text_avg - survey_score) >= {CONFLICT_GAP}, false) text_survey_conflict
  FROM sc2),
lb AS (
  SELECT *,
         max(CAST(n_detractor > 0 AS INT)) OVER l3 detr_l3, max(CAST(n_nps > 0 AS INT)) OVER l3 nps_l3,
         max(CAST(n_csat_low > 0 AS INT)) OVER l3 csat_low_l3, max(CAST(n_csat > 0 AS INT)) OVER l3 csat_l3,
         last_value(nps_last_in_month IGNORE NULLS) OVER upto last_nps_observed,
         last_value(CASE WHEN n_nps > 0 THEN mth END IGNORE NULLS) OVER upto last_nps_month,
         CAST(n_tx > 0 AS INT) + CAST(n_eng > 0 AS INT) + CAST(n_int > 0 AS INT) n_sources
  FROM cf
  WINDOW l3 AS (PARTITION BY customer_id ORDER BY mth ROWS BETWEEN {LOOKBACK_MONTHS - 1} PRECEDING AND CURRENT ROW),
         upto AS (PARTITION BY customer_id ORDER BY mth ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW)),
fin AS (
  SELECT *,
         CASE WHEN sentiment_score IS NULL THEN 'not_available'
              WHEN n_text > 0 AND survey_score IS NOT NULL AND NOT text_survey_conflict THEN 'high'
              WHEN survey_score IS NOT NULL THEN 'medium'
              WHEN n_text >= 2 THEN 'medium' ELSE 'low' END confidence_band,
         CASE WHEN sentiment_score IS NOT NULL THEN round(
              0.5 * (w_text_raw + w_csat_raw + w_nps_raw) / {WTOT}
            + 0.3 * least(n_text, 3) / 3.0
            + 0.2 * CASE WHEN n_text > 0 AND survey_score IS NOT NULL THEN 1 - abs(text_avg - survey_score) / 2 ELSE 0.5 END, 3)
         END confidence_score,
         CASE WHEN w_sum > 0 AND n_text > 0 THEN w_t * text_avg / w_sum END contrib_text,
         CASE WHEN w_sum > 0 AND n_csat > 0 THEN w_c * csat_norm_avg / w_sum END contrib_csat,
         CASE WHEN w_sum > 0 AND n_nps  > 0 THEN w_n * nps_norm_avg  / w_sum END contrib_nps
  FROM lb)
SELECT
  -- keys
  customer_id, mth, split,
  -- context only (never used in the score)
  loyalty_tier, customer_segment, region, preferred_channel, marketing_consent,
  -- activity coverage
  n_tx, n_eng, n_int, n_tx > 0 has_tx, n_eng > 0 has_eng, n_int > 0 has_int,
  (n_tx > 0 OR n_eng > 0 OR n_int > 0) active, n_sources,
  coalesce(nullif(concat_ws('+', CASE WHEN n_tx > 0 THEN 'TX' END, CASE WHEN n_eng > 0 THEN 'ENG' END,
                                 CASE WHEN n_int > 0 THEN 'INT' END), ''), 'none') source_combo,
  -- interaction evidence
  n_text, round(text_avg, 4) text_avg, round(text_min, 4) text_min, n_text_negative,
  n_issue_contacts, n_cancel_req, n_open_issue, last_interaction_ts,
  -- survey evidence (observed only; NULL when not observed)
  n_csat > 0 has_csat, n_csat, csat_avg, csat_min, csat_norm_avg, n_csat_low,
  n_nps > 0 has_nps, n_nps, nps_avg, nps_min, nps_max, nps_norm_avg, nps_first_in_month, nps_last_in_month,
  n_promoter, n_passive, n_detractor,
  CASE WHEN n_detractor > 0 THEN 'detractor' WHEN n_passive > 0 THEN 'passive' WHEN n_promoter > 0 THEN 'promoter' END nps_category_month,
  n_both_same_interaction, n_nps_on_open_issue,
  CASE WHEN n_csat > 0 AND n_nps > 0 THEN 'csat+nps' WHEN n_csat > 0 THEN 'csat_only'
       WHEN n_nps > 0 THEN 'nps_only' ELSE 'none' END survey_coverage,
  coalesce(nullif(concat_ws('+', CASE WHEN n_text > 0 THEN 'text' END, CASE WHEN n_csat > 0 THEN 'csat' END,
                                 CASE WHEN n_nps > 0 THEN 'nps' END), ''), 'none') evidence_basis,
  -- look-back context (three-state; NOT used in the score)
  CASE WHEN detr_l3 = 1 THEN 'detractor_seen' WHEN nps_l3 = 1 THEN 'surveyed_no_detractor' ELSE 'not_surveyed' END nps_status_l{LOOKBACK_MONTHS}m,
  CASE WHEN csat_low_l3 = 1 THEN 'low_csat_seen' WHEN csat_l3 = 1 THEN 'surveyed_no_low_csat' ELSE 'not_surveyed' END csat_status_l{LOOKBACK_MONTHS}m,
  last_nps_observed, last_nps_month, date_diff('month', last_nps_month, mth) months_since_last_nps,
  -- score
  CASE WHEN w_sum > 0 THEN round(w_t / w_sum, 4) END w_text_eff,
  CASE WHEN w_sum > 0 THEN round(w_c / w_sum, 4) END w_csat_eff,
  CASE WHEN w_sum > 0 THEN round(w_n / w_sum, 4) END w_nps_eff,
  round(contrib_text, 4) contrib_text, round(contrib_csat, 4) contrib_csat, round(contrib_nps, 4) contrib_nps,
  round(survey_score, 4) survey_score, round(sentiment_score, 4) sentiment_score,
  CASE WHEN sentiment_score IS NULL THEN 'not_available' ELSE 'scored' END sentiment_status,
  CASE WHEN sentiment_score IS NULL THEN NULL
       WHEN contrib_nps IS NOT NULL AND contrib_nps <= least(coalesce(contrib_text, 9), coalesce(contrib_csat, 9)) THEN 'nps'
       WHEN contrib_csat IS NOT NULL AND contrib_csat <= coalesce(contrib_text, 9) THEN 'csat'
       ELSE 'text' END lowest_component,     -- 9 = sentinel for argmin only, not a value
  nullif(concat_ws('; ',
     CASE WHEN n_detractor > 0 THEN 'NPS detractor (min ' || CAST(nps_min AS INT) || ')' END,
     CASE WHEN n_detractor = 0 AND n_promoter > 0 THEN 'NPS promoter (' || CAST(nps_max AS INT) || ')' END,
     CASE WHEN n_detractor = 0 AND n_promoter = 0 AND n_passive > 0 THEN 'NPS passive' END,
     CASE WHEN n_csat_low > 0 THEN 'Low CSAT (min ' || CAST(csat_min AS INT) || ')' END,
     CASE WHEN csat_avg >= 4.5 THEN 'High CSAT' END,
     CASE WHEN n_text_negative > 0 THEN 'Negative text on ' || n_text_negative || ' of ' || n_text || ' interaction(s)' END,
     CASE WHEN n_cancel_req > 0 THEN 'Cancellation request' END,
     CASE WHEN n_open_issue > 0 THEN n_open_issue || ' open/escalated issue(s)' END,
     CASE WHEN text_survey_conflict THEN 'Text and survey disagree' END), '') drivers,
  -- confidence
  text_survey_conflict, confidence_band, confidence_score,
  -- voice action (combined later with Behavioral Health)
  CASE WHEN sentiment_score IS NULL THEN 'no_voice_signal'
       WHEN n_cancel_req > 0 AND (n_detractor > 0 OR sentiment_score < 0) THEN 'retention_save_desk'
       WHEN n_detractor > 0 OR n_csat_low > 0 THEN 'service_recovery_48h'
       WHEN sentiment_score < -0.3 AND confidence_band IN ('high', 'medium') THEN 'proactive_service_check'
       WHEN n_promoter > 0 AND sentiment_score >= 0.5 THEN CASE WHEN marketing_consent THEN 'advocacy_referral_invite' ELSE 'thank_you_service_only' END
       WHEN n_passive > 0 THEN CASE WHEN marketing_consent THEN 'nurture_offer' ELSE 'monitor_no_marketing_consent' END
       ELSE 'monitor' END voice_action,
  -- lineage
  '{SCORE_VERSION}' score_version, '{json.dumps(WEIGHTS)}|cap={SURVEY_CAP}|gap={CONFLICT_GAP}|text={TEXT_SOURCE}' score_params,
  current_timestamp built_at
FROM fin
""")
con.execute(f"COPY cm_voice TO '{(GOLD_DIR / 'customer_month_voice.parquet').as_posix()}' (FORMAT parquet, COMPRESSION zstd)")
print(f"cm_voice: {q('SELECT count(*) FROM cm_voice').iloc[0,0]:,} rows in {time.time()-t0:,.0f}s")
display(q("SELECT * FROM cm_voice WHERE has_nps AND has_csat LIMIT 3").T)
````

### D4. Imputation guard tests (must all PASS)

````python
pyref = q("""SELECT n_text, text_avg, n_csat, csat_norm_avg, n_nps, nps_norm_avg, sentiment_score, survey_score,
                    confidence_band, confidence_score
             FROM cm_voice USING SAMPLE reservoir(200000 ROWS) REPEATABLE (7)""")
ref = composite(pyref.text_avg, pyref.n_text, pyref.csat_norm_avg, pyref.n_csat, pyref.nps_norm_avg, pyref.n_nps)
refc = confidence(pyref.n_text, pyref.text_avg, ref.survey_score, pyref.n_csat, pyref.n_nps)
diff = np.nanmax(np.abs(ref.score.values - pyref.sentiment_score.astype(float).values))
same_null = (np.isnan(ref.score.values) == pyref.sentiment_score.isna().values).all()

tests = {
 "NPS values NULL when no response":  "SELECT count(*) FROM cm_voice WHERE n_nps = 0 AND (nps_avg IS NOT NULL OR nps_norm_avg IS NOT NULL OR nps_min IS NOT NULL)",
 "CSAT values NULL when no response": "SELECT count(*) FROM cm_voice WHERE n_csat = 0 AND (csat_avg IS NOT NULL OR csat_norm_avg IS NOT NULL)",
 "NPS weight 0 when not observed":    "SELECT count(*) FROM cm_voice WHERE n_nps = 0 AND coalesce(w_nps_eff, 0) <> 0",
 "CSAT weight 0 when not observed":   "SELECT count(*) FROM cm_voice WHERE n_csat = 0 AND coalesce(w_csat_eff, 0) <> 0",
 "No score without evidence":         "SELECT count(*) FROM cm_voice WHERE n_text = 0 AND n_csat = 0 AND n_nps = 0 AND sentiment_score IS NOT NULL",
 "not_available <=> NULL score":      "SELECT count(*) FROM cm_voice WHERE (sentiment_score IS NULL) <> (confidence_band = 'not_available')",
 "Score within [-1, 1]":              "SELECT count(*) FROM cm_voice WHERE sentiment_score < -1 OR sentiment_score > 1",
 "Effective weights sum to 1":        "SELECT count(*) FROM cm_voice WHERE sentiment_score IS NOT NULL AND abs(w_text_eff + w_csat_eff + w_nps_eff - 1) > 1e-3",
 "Survey cap respected":              f"SELECT count(*) FROM cm_voice WHERE w_csat_eff + w_nps_eff > {SURVEY_CAP}/({WEIGHTS['text']}+{SURVEY_CAP}) + 1e-3 AND n_text > 0",
}
rows = [{"test": k, "violations": int(q(v).iloc[0, 0])} for k, v in tests.items()]
rows.append({"test": "SQL score == Python reference (200k sample)", "violations": int(not (diff < 1e-3 and same_null))})
rows.append({"test": "SQL confidence band == Python reference", "violations": int((refc.confidence_band.values != pyref.confidence_band.values).sum())})
rows.append({"test": "No profile column in score formula", "violations": sum(c in SCORE_SQL for c in PROFILE_COLS)})
rows.append({"test": "No look-back / last-NPS field in score formula",
             "violations": sum(c in SCORE_SQL for c in ("last_nps", "nps_status", "csat_status", "has_nps", "has_csat"))})
guard = pd.DataFrame(rows); guard["result"] = np.where(guard.violations == 0, "PASS", "FAIL")
save(guard, "D4_imputation_guard_tests", show=20)
assert (guard.result == "PASS").all(), "Imputation guard failed - fix before using the scores"
````

# Part E – Validation framework (NPS & sentiment)
**Principles:** (1) customer-grouped hold-out – every metric reported on **test customers**; (2) **no circularity** – a score that contains NPS is never validated against the same NPS (use *leave-NPS-out* scores or **future** outcomes); (3) report **n and coverage** with every metric; (4) report by **loyalty tier** to show fairness; (5) thresholds are *proposed* and must be agreed with the business.

| # | Check | Question it answers | Data | Metric(s) | Proposed pass |
|---|---|---|---|---|---|
| V0 | Split integrity | Is the hold-out clean? | all | customers/rows by split, overlap = 0 | overlap 0 |
| V1 | Text vs ground truth | Is the text component accurate? | test interactions; + unseen-template hold-out | MAE, Spearman, accuracy, macro-F1, weighted-F1 | macro-F1 ≥ 0.80 (cust.), ≥ 0.70 (template); ρ ≥ 0.70 |
| V2 | Text vs NPS/CSAT | Does text agree with what customers *say*? | test interactions with NPS | Spearman, detractor AUC, promoter AUC, 3-class agreement | ρ ≥ 0.50; AUC ≥ 0.75 |
| V3 | NPS incremental value | Does NPS add signal beyond text? | test NPS responses, future 3m | AUC(text) vs AUC(text+NPS) for future inactivity | Δ ≥ 0 |
| V4 | Weight tuning & ablation | Are weights justified? | survey months, future outcomes | grid on train → test; AUC inactivity, ρ with next NPS | default ≥ text-only |
| V5 | Calibration / monotonicity | Do higher scores mean better outcomes? | test scored months | decile ρ vs future inactivity; leave-NPS-out decile ρ vs NPS | ρ ≤ −0.80; ρ ≥ 0.80 |
| V6 | Confidence validity | Is confidence meaningful? | test months with NPS (leave-NPS-out) | error by band (low ≥ medium ≥ high) | monotonic |
| V7 | Missingness (MCAR) | Is non-imputation safe? | all interactions | response-rate spread; text mean with vs without NPS | ≤ 1.5 pp; ≤ 0.02 |
| V8 | Tier fairness | Does it work equally across tiers? | test, by loyalty tier | detractor AUC gap vs overall | ≤ 0.05 |
| V9 | Stability | Is it stable month to month? | all scored months | PSI vs reference month; lag-1 within-customer ρ | PSI ≤ 0.10 |
| V10 | Health cross-check (optional) | Does low Health precede detractors/inactivity? | Health table if present | detractor & inactivity rate by Health decile | monotonic |

````python
H = OUTCOME_HORIZON
con.execute(f"""
CREATE OR REPLACE TABLE val_cm AS
WITH o AS (
  SELECT customer_id, mth, split, loyalty_tier, n_int, n_text, text_avg, n_csat, csat_norm_avg, n_nps, nps_norm_avg, nps_avg,
         sentiment_score, sentiment_status, confidence_band, confidence_score,
         CASE WHEN mth <= DATE '{WINDOW_END}' - INTERVAL {H + 1} MONTH THEN
              1 - max(CAST(active AS INT)) OVER (PARTITION BY customer_id ORDER BY mth ROWS BETWEEN 1 FOLLOWING AND {H} FOLLOWING) END y_inactive_next,
         first_value(nps_first_in_month IGNORE NULLS) OVER (PARTITION BY customer_id ORDER BY mth
                                                            ROWS BETWEEN 1 FOLLOWING AND {H} FOLLOWING) y_next_nps
  FROM cm_voice)
SELECT * FROM o WHERE sentiment_status = 'scored'""")
V = q("SELECT * FROM val_cm")
print(f"validation customer-months (scored): {len(V):,}")
REPORT = []

def rec(check, metric, value, n=None, key=None, note=""):
    op, thr = THRESHOLDS.get(key, (None, None)) if key else (None, None)
    if op is None or value is None or (isinstance(value, float) and np.isnan(value)):
        status = "INFO"
    else:
        status = "PASS" if {">=": value >= thr, "<=": value <= thr, "==": value == thr}[op] else "FAIL"
    REPORT.append({"check": check, "metric": metric, "value": None if value is None else round(float(value), 4),
                   "n": n, "threshold": f"{op} {thr}" if op else "", "status": status, "note": note})

def auc(y, s):
    y, s = np.asarray(y), np.asarray(s)
    m = ~np.isnan(s.astype(float))
    return roc_auc_score(y[m], s[m]) if len(np.unique(y[m])) == 2 else np.nan
````

### V4. Weight tuning & ablation (non-circular: judged on **future** outcomes)
Grid on **train** survey-months (text weight fixed 0.6, no cap) → best by $J = 0.5\,AUC_{inactivity} + 0.5\,\frac{\rho_{next\ NPS}+1}{2}$ → compared on **test** with *text-only* and the *default* config.

````python
S = V[(V.n_text > 0) & ((V.n_csat > 0) | (V.n_nps > 0)) & V.y_inactive_next.notna()].copy()

def evaluate(df, w, cap):
    s = composite(df.text_avg, df.n_text, df.csat_norm_avg, df.n_csat, df.nps_norm_avg, df.n_nps, w=w, cap=cap).score.values
    a = auc(df.y_inactive_next.astype(int), -s)
    m = df.y_next_nps.notna().values
    rho = spearmanr(s[m], df.y_next_nps.values[m])[0] if m.sum() > 30 else np.nan
    return a, rho, 0.5 * a + 0.5 * (np.nan_to_num(rho) + 1) / 2

grid = []
for wc in (0, .1, .2, .3):
    for wn in (0, .1, .2, .3, .4):
        w = {"text": .6, "csat": wc, "nps": wn}
        a, rho, j = evaluate(S[S.split == "train"], w, None)
        grid.append({"w_csat": wc, "w_nps": wn, "auc_inactivity": a, "rho_next_nps": rho, "J": j})
grid = save(pd.DataFrame(grid).sort_values("J", ascending=False), "V4_weight_grid_train", show=8)
best = grid.iloc[0]
cfgs = {"text_only": ({"text": 1, "csat": 0, "nps": 0}, None),
        "default": (WEIGHTS, SURVEY_CAP),
        "best_grid": ({"text": .6, "csat": best.w_csat, "nps": best.w_nps}, None)}
out = []
for k, (w, cap) in cfgs.items():
    a, rho, j = evaluate(S[S.split == "test"], w, cap)
    out.append({"config": k, "weights": json.dumps(w), "cap": cap, "auc_inactivity_test": a, "rho_next_nps_test": rho, "J_test": j})
v4 = save(pd.DataFrame(out), "V4_config_comparison_test")
d = v4.set_index("config")
rec("V4", "AUC inactivity: default - text-only (test)", d.loc["default", "auc_inactivity_test"] - d.loc["text_only", "auc_inactivity_test"],
    int((S.split == "test").sum()), "V4_delta_auc_default_vs_text")
rec("V4", "Spearman with next NPS: default (test)", d.loc["default", "rho_next_nps_test"])
rec("V4", "Best grid weights (train)", None, note=f"csat={best.w_csat}, nps={best.w_nps}; J_test={d.loc['best_grid','J_test']:.4f} "
                                                  f"vs default {d.loc['default','J_test']:.4f}")
````

### V8. Fairness by loyalty tier (context only – never in the score)

````python
rows = []
overall = auc((t2.nps <= 6).astype(int), -t2.text_pred)
rng = np.random.default_rng(42)
def boot_ci(y, s, reps=200):
    y, s = np.asarray(y), np.asarray(s); vals = []
    for _ in range(reps):
        i = rng.integers(0, len(y), len(y))
        if len(np.unique(y[i])) == 2:
            vals.append(roc_auc_score(y[i], s[i]))
    return np.percentile(vals, [2.5, 97.5])
for tier, g in t2.groupby("loyalty_tier"):
    vt = V[(V.split == "test") & (V.loyalty_tier == tier) & V.y_inactive_next.notna()]
    rows.append({"loyalty_tier": tier, "n_nps": len(g), "spearman_text_nps": spearmanr(g.text_pred, g.nps)[0],
                 "auc_detractor": auc((g.nps <= 6).astype(int), -g.text_pred),
                 "auc_ci_low": boot_ci((g.nps <= 6).astype(int), -g.text_pred)[0],
                 "auc_ci_high": boot_ci((g.nps <= 6).astype(int), -g.text_pred)[1],
                 "detractor_rate": (g.nps <= 6).mean(),
                 "auc_inactivity_score": auc(vt.y_inactive_next.astype(int), -vt.sentiment_score),
                 "avg_sentiment_score": vt.sentiment_score.mean(), "avg_nps": g.nps.mean()})
v8 = save(pd.DataFrame(rows), "V8_tier_fairness")
worst = v8.loc[(v8.auc_detractor - overall).abs().idxmax()]
rec("V8", "Max |detractor AUC tier - overall|", (v8.auc_detractor - overall).abs().max(), len(t2), "V8_max_tier_auc_gap",
    f"worst: {worst.loyalty_tier} AUC {worst.auc_detractor:.3f} (95% CI {worst.auc_ci_low:.3f}-{worst.auc_ci_high:.3f}, n={worst.n_nps}, "
    f"detractor rate {worst.detractor_rate:.1%}) vs overall {overall:.3f}")
````

### V9. Stability over time

````python
def psi(ref, cur, bins):
    e = np.histogram(ref, bins)[0] / len(ref); a = np.histogram(cur, bins)[0] / len(cur)
    e, a = np.clip(e, 1e-6, None), np.clip(a, 1e-6, None)
    return float(np.sum((a - e) * np.log(a / e)))

months = sorted(V.mth.unique()); ref_m = months[0]
ref = V.loc[V.mth == ref_m, "sentiment_score"].values
bins = np.unique(np.quantile(ref, np.linspace(0, 1, 11))); bins[0], bins[-1] = -1.0001, 1.0001
v9 = pd.DataFrame([{"mth": m, "n": int((V.mth == m).sum()), "mean_score": V.loc[V.mth == m, "sentiment_score"].mean(),
                    "psi_vs_ref": psi(ref, V.loc[V.mth == m, "sentiment_score"].values, bins)} for m in months])
save(v9, "V9_stability_psi")
rec("V9", f"Max PSI vs {pd.Timestamp(ref_m):%Y-%m}", v9.psi_vs_ref.max(), len(V), "V9_max_psi")
s = V[["customer_id", "mth", "sentiment_score"]].sort_values(["customer_id", "mth"])
s["prev"] = s.groupby("customer_id").sentiment_score.shift(); s["gap"] = s.groupby("customer_id").mth.diff().dt.days
c1 = s[(s.gap > 0) & (s.gap <= 31)]
rec("V9", "Lag-1 within-customer correlation (consecutive months)", c1.sentiment_score.corr(c1.prev), len(c1),
    note="low values = score is event-driven (expected for interaction-based voice)")
````

# Part F – Behavioural feature layer for the survey analysis (shared by NPS and CSAT)
**Why this part exists (NPS_testing.docx §1–§3):** Parts A–E relate NPS/CSAT mainly to *interaction* attributes. The customer-month
model only *counts* transactions and engagements (`n_tx`, `n_eng`, `active`, `n_sources`, `source_combo`) to decide whether a customer
was active – no direct relationship between the surveys and transaction or engagement behaviour is tested, and future inactivity is the
combined "no transaction AND no engagement AND no interaction" outcome (so the A9 detractor rate cannot be read as *transaction* inactivity).

This part builds the features that Parts G (NPS), H (CSAT) and I (comparison) need, **once**, so both metrics use identical inputs:

| Step | Output | Grain |
|---|---|---|
| F0 | configuration: column candidates, business mappings, survey specs | – |
| F1 | column resolution + category/mapping check against the actual data | – |
| F2 | `tx_cm` – transaction features (count, payments, failures, refunds, reversals, spend in own currency + within-currency percentile) | customer × month |
| F3 | `eng_cm` – engagement features (events, logins, app/web, email, campaign, conversion, session, pages, opt-outs, per event type) | customer × month |
| F4 | `beh_cm` – spine from `cm_voice` + trailing trend, prior window and **separate** future outcomes (no-TX / no-ENG / no-INT / fully inactive) | customer × month |
| F5 | `srv_resp` – one row per survey response with exact recency (ASOF) and the customer-month behaviour joined 1:1 | interaction (response) |

**Methodological rule (doc §4.3):** survey responses are the starting observations. Transactions and engagements are first aggregated to
customer × month and only then joined on `(customer_id, mth)` – raw events are never joined to interactions, so survey records cannot be duplicated (guard test in F5).
**Golden rule unchanged:** CSAT/NPS are never imputed; every analysis uses observed responses only. Behavioural *counts* of 0 are facts, not imputations;
behavioural *values* (spend, session length, recency) stay NULL when there is no underlying event.

### F1. Column resolution and category / mapping check (verify mappings against the data before using them)

````python
def norm_sql(col):
    """SQL expression that normalises a categorical value (NULL-safe)."""
    return f"lower(regexp_replace(trim(CAST({col} AS VARCHAR)), '[ -]+', '_', 'g'))"

def resolve_cols(view, candidates):
    """Map logical names -> actual column names of `view` (None when not present)."""
    if not table_exists(view):
        return {k: None for k in candidates}
    actual = {c.lower(): c for c in q(f"DESCRIBE {view}").column_name}
    return {k: next((actual[c.lower()] for c in cands if c.lower() in actual), None) for k, cands in candidates.items()}

def bool_sql(col):
    return f"(upper(trim(CAST({col} AS VARCHAR))) IN ('Y','YES','TRUE','1'))"

TXC = resolve_cols("raw_transactions", TX_COL_CANDIDATES) if HAS_TX else {k: None for k in TX_COL_CANDIDATES}
ENC = resolve_cols("raw_engagements", ENG_COL_CANDIDATES) if HAS_ENG else {k: None for k in ENG_COL_CANDIDATES}
ITC = resolve_cols("raw_interactions", ITX_COL_CANDIDATES)
col_map = pd.DataFrame([{"dataset": d, "logical": k, "resolved_column": v or "-- not found --"}
                        for d, m in (("transactions", TXC), ("engagements", ENC), ("interactions", ITC)) for k, v in m.items()])
save(col_map, "F1_column_resolution", show=30)
missing = col_map[col_map.resolved_column == "-- not found --"]
if len(missing):
    flag("F1", "Columns not found (features depending on them are left NULL and skipped): "
               + ", ".join(f"{r.dataset}.{r.logical}" for r in missing.itertuples()))

freq_parts = []
for ds, view, m, keys in (("transactions", "raw_transactions", TXC, ("type", "status", "currency")),
                          ("engagements", "raw_engagements", ENC, ("channel", "event"))):
    for k in keys:
        if m.get(k):
            freq_parts.append(q(f"""SELECT '{ds}' dataset, '{k}' field, {norm_sql(m[k])} value, count(*) n
                                    FROM {view} GROUP BY ALL""").assign(pct=lambda d: (100 * d.n / d.n.sum()).round(3)))
freq = pd.concat(freq_parts, ignore_index=True) if freq_parts else pd.DataFrame(columns=["dataset", "field", "value", "n", "pct"])
save(freq.sort_values(["dataset", "field", "n"], ascending=[True, True, False]), "F1_category_frequencies", show=0)

MAPPINGS = {("transactions", "type"): TX_PAYMENT_TYPES + TX_REFUND_TYPES,
            ("transactions", "status"): TX_SUCCESS_STATUS + TX_FAILED_STATUS + TX_REVERSED_STATUS,
            ("engagements", "channel"): ENG_APP_CHANNELS + ENG_WEB_CHANNELS + ENG_EMAIL_CHANNELS,
            ("engagements", "event"): ENG_LOGIN_EVENTS + ENG_EMAIL_OPEN_EVENTS + ENG_EMAIL_CLICK_EVENTS + ENG_OPTOUT_EVENTS + ENG_RESPONSE_EVENTS}
mrows = []
for (ds, field), vals in MAPPINGS.items():
    seen = set(freq[(freq.dataset == ds) & (freq.field == field)].value)
    for v in dict.fromkeys(vals):
        mrows.append({"dataset": ds, "field": field, "configured_value": v, "found_in_data": v in seen,
                      "n": int(freq.loc[(freq.dataset == ds) & (freq.field == field) & (freq.value == v), "n"].sum())})
mcheck = save(pd.DataFrame(mrows), "F1_mapping_check", show=0)
nf = mcheck[~mcheck.found_in_data]
if len(nf):
    flag("F1", f"{len(nf)} configured mapping value(s) not found in the data: "
               + ", ".join(f"{r.field}={r.configured_value}" for r in nf.itertuples()) + " -> review F0 mappings")
else:
    print("all configured mapping values found in the data")
````

### F2. Transaction features per customer-month (`tx_cm`)
* Payments = `TX_PAYMENT_TYPES`; successful / failed are split by status; reversals and completed refunds are counted separately
  (doc §1.3: refunds, payment failures and successful transactions are different signals).
* **Currencies are never added together.** Amounts are kept in the customer-month's *primary* currency (most transactions) and
  compared through a **within-currency, within-month percentile (0–100)** – the cross-customer comparable measure.
* `days_since_last_tx` is computed exactly per survey response in F5 (ASOF join); here only the month's last timestamp is kept.

````python
def sql_or_null(expr, *needs, cast="DOUBLE"):
    """Return `expr` when every required source column was resolved, else a typed NULL (feature unavailable, not zero)."""
    return expr if all(needs) else f"CAST(NULL AS {cast})"

if HAS_TX and TXC["ts"]:
    t_type = norm_sql(TXC["type"]) if TXC["type"] else "NULL"
    t_stat = norm_sql(TXC["status"]) if TXC["status"] else "NULL"
    t_amt = f"TRY_CAST({TXC['amount']} AS DOUBLE)" if TXC["amount"] else "NULL"
    t_cur = f"upper(trim(CAST({TXC['currency']} AS VARCHAR)))" if TXC["currency"] else "'N/A'"
    con.execute(f"""
    CREATE OR REPLACE TABLE tx_ev AS
    SELECT * FROM (SELECT customer_id, TRY_CAST({TXC['ts']} AS TIMESTAMP) ts, {t_type} ttype, {t_stat} tstatus,
                          {t_amt} amount, {t_cur} currency FROM raw_transactions)
    WHERE ts >= TIMESTAMP '{WINDOW_START}' AND ts < TIMESTAMP '{WINDOW_END}'""")
    PAY, SUC, FAIL_, REV, REF = (sql_list(TX_PAYMENT_TYPES), sql_list(TX_SUCCESS_STATUS), sql_list(TX_FAILED_STATUS),
                                 sql_list(TX_REVERSED_STATUS), sql_list(TX_REFUND_TYPES))
    ty, st, am = TXC["type"], TXC["status"], TXC["amount"]
    con.execute(f"""
    CREATE OR REPLACE TABLE tx_cm AS
    WITH a AS (
      SELECT customer_id, CAST(date_trunc('month', ts) AS DATE) mth, currency, count(*) n_cur, max(ts) last_ts,
             {sql_or_null(f"count_if(ttype IN {PAY})", ty, cast='BIGINT')} n_pay_attempts,
             {sql_or_null(f"count_if(ttype IN {PAY} AND tstatus IN {SUC})", ty, st, cast='BIGINT')} n_pay_success,
             {sql_or_null(f"count_if(ttype IN {PAY} AND tstatus IN {FAIL_})", ty, st, cast='BIGINT')} n_pay_failed,
             {sql_or_null(f"count_if(tstatus IN {REV})", st, cast='BIGINT')} n_reversed,
             {sql_or_null(f"count_if(ttype IN {REF} AND tstatus IN {SUC})", ty, st, cast='BIGINT')} n_refund,
             {sql_or_null(f"sum(abs(amount)) FILTER (WHERE ttype IN {REF} AND tstatus IN {SUC})", ty, st, am)} refund_amount,
             {sql_or_null(f"sum(amount) FILTER (WHERE ttype IN {PAY} AND tstatus IN {SUC})", ty, st, am)} gross_spend,
             {sql_or_null(f"avg(amount) FILTER (WHERE ttype IN {PAY} AND tstatus IN {SUC})", ty, st, am)} avg_tx_value
      FROM tx_ev GROUP BY ALL),
    r AS (SELECT *, row_number() OVER (PARTITION BY customer_id, mth ORDER BY n_cur DESC, currency) rn FROM a),
    c AS (
      SELECT customer_id, mth, count(*) n_currencies, max(last_ts) last_tx_ts_in_month,
             sum(n_pay_attempts) n_pay_attempts, sum(n_pay_success) n_pay_success, sum(n_pay_failed) n_pay_failed,
             sum(n_reversed) n_reversed, sum(n_refund) n_refund,
             max(CASE WHEN rn = 1 THEN currency END) primary_currency,
             max(CASE WHEN rn = 1 THEN gross_spend END) gross_spend,       -- primary currency only (never mixed)
             max(CASE WHEN rn = 1 THEN avg_tx_value END) avg_tx_value,
             max(CASE WHEN rn = 1 THEN refund_amount END) refund_amount
      FROM r GROUP BY ALL)
    SELECT *,
           n_pay_failed / nullif(n_pay_attempts, 0) pay_failure_rate,
           CASE WHEN gross_spend IS NOT NULL THEN round(100 * percent_rank() OVER
                (PARTITION BY primary_currency, mth, gross_spend IS NULL ORDER BY gross_spend), 2) END gross_spend_pctile,
           CASE WHEN avg_tx_value IS NOT NULL THEN round(100 * percent_rank() OVER
                (PARTITION BY primary_currency, mth, avg_tx_value IS NULL ORDER BY avg_tx_value), 2) END avg_tx_value_pctile,
           CASE WHEN refund_amount IS NOT NULL THEN round(100 * percent_rank() OVER
                (PARTITION BY primary_currency, mth, refund_amount IS NULL ORDER BY refund_amount), 2) END refund_amount_pctile
    FROM c""")
else:
    con.execute("""CREATE OR REPLACE TABLE tx_ev AS SELECT CAST(NULL AS VARCHAR) customer_id, CAST(NULL AS TIMESTAMP) ts WHERE false""")
    con.execute("""CREATE OR REPLACE TABLE tx_cm AS SELECT CAST(NULL AS VARCHAR) customer_id, CAST(NULL AS DATE) mth WHERE false""")
    flag("F2", "transactions not available -> transaction features skipped")
display(q("SELECT count(*) customer_months, count(DISTINCT customer_id) customers, sum(n_pay_failed) failed, sum(n_refund) refunds "
          "FROM tx_cm") if HAS_TX and TXC["ts"] else q("SELECT 0 customer_months"))
````

### F3. Engagement features per customer-month (`eng_cm`)
Login, app, web, email open/click, campaign response, conversion, session duration (non-zero sessions only), pages viewed, opt-outs,
plus one count column per observed event type (doc §2.1 "NPS by engagement type").

````python
EVT_COLS = []
if HAS_ENG and ENC["ts"]:
    e_ch = norm_sql(ENC["channel"]) if ENC["channel"] else "NULL"
    e_ev = norm_sql(ENC["event"]) if ENC["event"] else "NULL"
    e_du = f"TRY_CAST({ENC['duration']} AS DOUBLE)" if ENC["duration"] else "NULL"
    e_pg = f"TRY_CAST({ENC['pages']} AS DOUBLE)" if ENC["pages"] else "NULL"
    e_cp = f"nullif(trim(CAST({ENC['campaign']} AS VARCHAR)), '')" if ENC["campaign"] else "NULL"
    e_cv = bool_sql(ENC["converted"]) if ENC["converted"] else "NULL"
    con.execute(f"""
    CREATE OR REPLACE TABLE eng_ev AS
    SELECT * FROM (SELECT customer_id, TRY_CAST({ENC['ts']} AS TIMESTAMP) ts, {e_ch} channel, {e_ev} event,
                          {e_du} duration, {e_pg} pages, {e_cp} campaign, {e_cv} converted FROM raw_engagements)
    WHERE ts >= TIMESTAMP '{WINDOW_START}' AND ts < TIMESTAMP '{WINDOW_END}'""")
    if ENC["event"]:
        top_ev = q(f"SELECT event, count(*) n FROM eng_ev WHERE event IS NOT NULL GROUP BY 1 ORDER BY n DESC LIMIT {MAX_EVENT_TYPE_COLS}").event
        EVT_COLS = [(e, "n_evt_" + "".join(ch if ch.isalnum() else "_" for ch in e)) for e in top_ev]
    evt_sql = "".join(f", count_if(event = '{e}') {c}" for e, c in EVT_COLS)
    ch_, ev_, du_, pg_, cp_, cv_ = ENC["channel"], ENC["event"], ENC["duration"], ENC["pages"], ENC["campaign"], ENC["converted"]
    resp_cond = "(" + " OR ".join(x for x in (f"event IN {sql_list(ENG_RESPONSE_EVENTS)}" if ev_ else None,
                                              "converted" if cv_ else None) if x) + ")" if (ev_ or cv_) else None
    con.execute(f"""
    CREATE OR REPLACE TABLE eng_cm AS
    SELECT customer_id, CAST(date_trunc('month', ts) AS DATE) mth, count(*) n_eng_events, max(ts) last_eng_ts_in_month,
           {sql_or_null(f"count_if(event IN {sql_list(ENG_LOGIN_EVENTS)})", ev_, cast='BIGINT')} n_login,
           {sql_or_null(f"count_if(channel IN {sql_list(ENG_APP_CHANNELS)})", ch_, cast='BIGINT')} n_app_events,
           {sql_or_null(f"count_if(channel IN {sql_list(ENG_WEB_CHANNELS)})", ch_, cast='BIGINT')} n_web_events,
           {sql_or_null(f"count_if(channel IN {sql_list(ENG_EMAIL_CHANNELS)})", ch_, cast='BIGINT')} n_email_events,
           {sql_or_null(f"count_if(event IN {sql_list(ENG_EMAIL_OPEN_EVENTS)})", ev_, cast='BIGINT')} n_email_open,
           {sql_or_null(f"count_if(event IN {sql_list(ENG_EMAIL_CLICK_EVENTS)})", ev_, cast='BIGINT')} n_email_click,
           {sql_or_null(f"count_if(event IN {sql_list(ENG_OPTOUT_EVENTS)})", ev_, cast='BIGINT')} n_optout,
           {sql_or_null("count_if(campaign IS NOT NULL)", cp_, cast='BIGINT')} n_campaign_events,
           {sql_or_null(f"count_if(campaign IS NOT NULL AND {resp_cond})", cp_, resp_cond, cast='BIGINT')} n_campaign_responses,
           {sql_or_null("count_if(converted)", cv_, cast='BIGINT')} n_conversions,
           {sql_or_null("sum(duration) FILTER (WHERE duration > 0)", du_)} total_session_sec,
           {sql_or_null("avg(duration) FILTER (WHERE duration > 0)", du_)} avg_session_sec,     -- non-zero sessions only
           {sql_or_null("sum(pages)", pg_)} n_pages
           {evt_sql}
    FROM eng_ev GROUP BY ALL""")
    con.execute("""CREATE OR REPLACE TABLE eng_cm AS
                   SELECT *, n_email_click / nullif(n_email_open, 0) email_click_to_open,
                             n_campaign_responses / nullif(n_campaign_events, 0) campaign_response_rate FROM eng_cm""")
else:
    con.execute("""CREATE OR REPLACE TABLE eng_ev AS SELECT CAST(NULL AS VARCHAR) customer_id, CAST(NULL AS TIMESTAMP) ts WHERE false""")
    con.execute("""CREATE OR REPLACE TABLE eng_cm AS SELECT CAST(NULL AS VARCHAR) customer_id, CAST(NULL AS DATE) mth WHERE false""")
    flag("F3", "engagements not available -> engagement features skipped")
print("event-type columns:", [c for _, c in EVT_COLS])
````

### F4. Customer-month behaviour table (`beh_cm`, gold) – trailing trend, prior window and separated future outcomes
* Spine = `cm_voice` (every eligible month from signup), so `ROWS` windows are calendar months.
* **Trailing trend** (known at survey time): current month vs average of the previous `OUTCOME_HORIZON` months →
  `declining / stable / increasing / no_prior_activity / insufficient_history`.
* **Future outcomes** (next `OUTCOME_HORIZON` months, only where the full horizon is observed) – kept **separate** (doc §1.2, §3.1):
  `y_no_tx_next`, `y_no_eng_next`, `y_no_int_next`, `y_inactive_next` (= all three, same definition as A9/V3), and
  `y_tx_decline_next` / `y_eng_decline_next` (future window < (1−`DECLINE_PCT`) × prior window of the same length).

````python
H = OUTCOME_HORIZON
tx_cols = [c for c in q("DESCRIBE tx_cm").column_name if c not in ("customer_id", "mth")]
eng_cols = [c for c in q("DESCRIBE eng_cm").column_name if c not in ("customer_id", "mth")]
def available(table, cols):
    """Columns that hold at least one non-NULL value (i.e. their source column was resolved)."""
    if not cols:
        return set()
    r = q(f"SELECT {', '.join(f'count({c}) {c}' for c in cols)} FROM {table}").iloc[0]
    return {c for c in cols if r[c] > 0}
# Event COUNTS of a month without any transaction/engagement are a fact (0), exactly like n_tx / n_eng in cm_voice.
# Values, rates and percentiles (spend, avg session, failure rate, ...) stay NULL - they do not exist without events.
COUNT_FILL = {c for c in available("tx_cm", tx_cols) | available("eng_cm", eng_cols) if c.startswith(("n_", "total_"))}
def beh_col(alias, c):
    return f", coalesce({alias}.{c}, 0) {c}" if c in COUNT_FILL else f", {alias}.{c}"
def trend_sql(cur, prior, n_prior):
    return f"""CASE WHEN {n_prior} < {H} THEN 'insufficient_history'
                    WHEN {prior} = 0 AND {cur} = 0 THEN 'no_prior_activity'
                    WHEN {prior} = 0 THEN 'new_activity'
                    WHEN {cur} < (1 - {DECLINE_PCT}) * {prior} THEN 'declining'
                    WHEN {cur} > (1 + {DECLINE_PCT}) * {prior} THEN 'increasing' ELSE 'stable' END"""
con.execute(f"""
CREATE OR REPLACE TABLE beh_cm AS
WITH b AS (
  SELECT v.customer_id, v.mth, v.split, v.loyalty_tier, v.customer_segment, v.region, v.active, v.source_combo, v.n_sources,
         v.n_tx, v.n_eng, v.n_int, v.has_tx, v.has_eng, v.has_int, v.n_text, v.text_avg, v.n_text_negative,
         v.n_issue_contacts, v.n_open_issue, v.n_cancel_req
         {"".join(beh_col("t", c) for c in tx_cols)} {"".join(beh_col("e", c) for c in eng_cols)}
  FROM cm_voice v LEFT JOIN tx_cm t USING (customer_id, mth) LEFT JOIN eng_cm e USING (customer_id, mth)),
w AS (
  SELECT *,
         count(*) OVER prv n_prior_months,
         avg(n_tx) OVER prv n_tx_prior_avg, avg(n_eng) OVER prv n_eng_prior_avg,
         sum(n_tx) OVER prv n_tx_prior, sum(n_eng) OVER prv n_eng_prior, sum(n_int) OVER prv n_int_prior,
         n_tx - lag(n_tx) OVER o tx_count_mom_change, n_eng - lag(n_eng) OVER o eng_count_mom_change,
         lag(gross_spend) OVER o prev_gross_spend, lag(primary_currency) OVER o prev_currency,
         sum(n_tx) OVER nxt fut_n_tx, sum(n_eng) OVER nxt fut_n_eng, sum(n_int) OVER nxt fut_n_int,
         sum(CAST(active AS INT)) OVER nxt fut_active_months
  FROM b
  WINDOW o AS (PARTITION BY customer_id ORDER BY mth),
         prv AS (PARTITION BY customer_id ORDER BY mth ROWS BETWEEN {H} PRECEDING AND 1 PRECEDING),
         nxt AS (PARTITION BY customer_id ORDER BY mth ROWS BETWEEN 1 FOLLOWING AND {H} FOLLOWING))
SELECT * EXCLUDE (prev_gross_spend, prev_currency),
       CASE WHEN prev_gross_spend > 0 AND gross_spend IS NOT NULL AND prev_currency = primary_currency
            THEN round((gross_spend - prev_gross_spend) / prev_gross_spend, 4) END spend_mom_pct,   -- same currency only
       CASE WHEN n_prior_months = {H} AND n_tx_prior_avg > 0 THEN round(n_tx / n_tx_prior_avg, 4) END tx_trend_ratio,
       CASE WHEN n_prior_months = {H} AND n_eng_prior_avg > 0 THEN round(n_eng / n_eng_prior_avg, 4) END eng_trend_ratio,
       {trend_sql('n_tx', 'n_tx_prior_avg', 'n_prior_months')} tx_trend,
       {trend_sql('n_eng', 'n_eng_prior_avg', 'n_prior_months')} eng_trend,
       mth <= DATE '{WINDOW_END}' - INTERVAL {H + 1} MONTH has_full_horizon,
       CASE WHEN mth <= DATE '{WINDOW_END}' - INTERVAL {H + 1} MONTH THEN CAST(fut_n_tx = 0 AS INT) END y_no_tx_next,
       CASE WHEN mth <= DATE '{WINDOW_END}' - INTERVAL {H + 1} MONTH THEN CAST(fut_n_eng = 0 AS INT) END y_no_eng_next,
       CASE WHEN mth <= DATE '{WINDOW_END}' - INTERVAL {H + 1} MONTH THEN CAST(fut_n_int = 0 AS INT) END y_no_int_next,
       CASE WHEN mth <= DATE '{WINDOW_END}' - INTERVAL {H + 1} MONTH THEN CAST(fut_active_months = 0 AS INT) END y_inactive_next,
       CASE WHEN mth <= DATE '{WINDOW_END}' - INTERVAL {H + 1} MONTH AND n_prior_months = {H} AND n_tx_prior > 0
            THEN CAST(fut_n_tx < (1 - {DECLINE_PCT}) * n_tx_prior AS INT) END y_tx_decline_next,
       CASE WHEN mth <= DATE '{WINDOW_END}' - INTERVAL {H + 1} MONTH AND n_prior_months = {H} AND n_eng_prior > 0
            THEN CAST(fut_n_eng < (1 - {DECLINE_PCT}) * n_eng_prior AS INT) END y_eng_decline_next
FROM w""")
con.execute(f"COPY beh_cm TO '{(GOLD_DIR / 'customer_month_behaviour.parquet').as_posix()}' (FORMAT parquet, COMPRESSION zstd)")
chk = q("""SELECT count(*) rows_, count(DISTINCT (customer_id, mth)) keys_,
                  (SELECT count(*) FROM cm_voice) cm_voice_rows,
                  count_if(y_inactive_next = 1 AND (y_no_tx_next = 0 OR y_no_eng_next = 0 OR y_no_int_next = 0)) inconsistent_outcomes
           FROM beh_cm""")
display(chk)
assert chk.rows_.iloc[0] == chk.keys_.iloc[0] == chk.cm_voice_rows.iloc[0], "beh_cm must be 1 row per cm_voice customer-month"
assert chk.inconsistent_outcomes.iloc[0] == 0, "y_inactive_next must imply no TX, no ENG and no INT"
save(q("""SELECT count_if(has_full_horizon AND active) active_cm_with_horizon,
                 round(100*avg(y_no_tx_next) FILTER (WHERE active), 2) pct_no_tx_next,
                 round(100*avg(y_no_eng_next) FILTER (WHERE active), 2) pct_no_eng_next,
                 round(100*avg(y_no_int_next) FILTER (WHERE active), 2) pct_no_int_next,
                 round(100*avg(y_inactive_next) FILTER (WHERE active), 2) pct_fully_inactive_next
          FROM beh_cm"""), "F4_future_outcome_base_rates")
````

### F5. Survey response fact (`srv_resp`, silver) – one row per interaction with a CSAT and/or NPS response
* Exact **recency at survey time** via DuckDB `ASOF JOIN` (last transaction / engagement strictly before the response timestamp;
  NULL = none earlier in the window, i.e. left-censored – not zero).
* **Repeat contact** = another interaction by the same customer in the `REPEAT_CONTACT_DAYS` days before.
* Behaviour joined 1:1 (LEFT JOIN) on `(customer_id, mth)` from `beh_cm` → guard: row count unchanged (no duplication, no loss).

````python
itx_dur = (f"LEFT JOIN (SELECT interaction_id, TRY_CAST({ITC['duration']} AS DOUBLE) resolution_time_sec "
           f"FROM raw_interactions QUALIFY row_number() OVER (PARTITION BY interaction_id) = 1) d USING (interaction_id)"
           if ITC["duration"] else "")
itx_dur_col = "d.resolution_time_sec" if ITC["duration"] else "CAST(NULL AS DOUBLE) resolution_time_sec"
con.execute(f"""
CREATE OR REPLACE TABLE srv_base AS
WITH f AS (
  SELECT f.*, count(*) OVER (PARTITION BY customer_id ORDER BY ts
                             RANGE BETWEEN INTERVAL {REPEAT_CONTACT_DAYS} DAYS PRECEDING AND CURRENT ROW) - 1 n_prior_contacts_{REPEAT_CONTACT_DAYS}d
  FROM fis f)
SELECT f.interaction_id, f.customer_id, f.ts, f.mth, f.split, f.channel, f.interaction_type, f.resolution_status, f.fcr,
       f.text_pred, f.text_label_pred, f.gt_label, f.csat, f.nps, f.csat_norm, f.nps_norm,
       f.n_prior_contacts_{REPEAT_CONTACT_DAYS}d, CAST(f.n_prior_contacts_{REPEAT_CONTACT_DAYS}d > 0 AS INT) repeat_contact,
       {itx_dur_col}
FROM f {itx_dur}
WHERE f.csat IS NOT NULL OR f.nps IS NOT NULL""")
n_base = q("SELECT count(*) FROM srv_base").iloc[0, 0]
con.execute(f"""
CREATE OR REPLACE TABLE srv_resp AS
WITH r1 AS (SELECT s.*, date_diff('day', t.ts, s.ts) days_since_last_tx
            FROM srv_base s ASOF LEFT JOIN (SELECT customer_id, ts FROM tx_ev) t
                 ON s.customer_id = t.customer_id AND s.ts > t.ts),
     r2 AS (SELECT r1.*, date_diff('day', e.ts, r1.ts) days_since_last_eng
            FROM r1 ASOF LEFT JOIN (SELECT customer_id, ts FROM eng_ev) e
                 ON r1.customer_id = e.customer_id AND r1.ts > e.ts)
SELECT r2.*, b.* EXCLUDE (customer_id, mth, split)
FROM r2 LEFT JOIN beh_cm b USING (customer_id, mth)""")
n_resp, n_no_spine = (int(x) for x in q("SELECT count(*), count_if(active IS NULL) FROM srv_resp").iloc[0])
assert n_resp == n_base, f"duplicate/lost survey records after the behaviour join ({n_base:,} -> {n_resp:,})"
if n_no_spine:
    flag("F5", f"{n_no_spine:,} survey responses fall in months outside the customer spine (before signup month) - "
               "kept, but their behaviour features are NULL and they are excluded from outcome analyses")
con.execute(f"COPY srv_resp TO '{(SILVER_DIR / 'survey_response_behaviour.parquet').as_posix()}' (FORMAT parquet, COMPRESSION zstd)")
RESP = q("SELECT * FROM srv_resp")
RESP["has_full_horizon"] = RESP.has_full_horizon.fillna(False).astype(bool)
print(f"srv_resp: {n_resp:,} survey responses (no duplication) | with NPS {RESP.nps.notna().sum():,} | with CSAT {RESP.csat.notna().sum():,}")

# Feature catalogue used by Parts G, H and I (features whose source columns were not found are dropped automatically)
FEATURES = {
    "transactions": ["n_tx", "n_pay_attempts", "n_pay_success", "n_pay_failed", "pay_failure_rate", "n_reversed", "n_refund",
                     "refund_amount_pctile", "gross_spend_pctile", "avg_tx_value_pctile", "days_since_last_tx",
                     "tx_count_mom_change", "spend_mom_pct", "tx_trend_ratio"],
    "engagements":  ["n_eng", "n_login", "n_app_events", "n_web_events", "n_email_open", "n_email_click", "email_click_to_open",
                     "n_campaign_responses", "campaign_response_rate", "n_conversions", "avg_session_sec", "total_session_sec",
                     "n_pages", "n_optout", "days_since_last_eng", "eng_count_mom_change", "eng_trend_ratio"],
    "interactions": ["n_int", "n_issue_contacts", "n_open_issue", "n_cancel_req", "repeat_contact",
                     f"n_prior_contacts_{REPEAT_CONTACT_DAYS}d", "resolution_time_sec", "text_pred"],
}
FEATURES = {ds: [c for c in cols if c in RESP.columns and RESP[c].notna().any()] for ds, cols in FEATURES.items()}
FEATURE_DATASET = {c: ds for ds, cols in FEATURES.items() for c in cols}
for c in FEATURE_DATASET:
    RESP[c] = pd.to_numeric(RESP[c], errors="coerce").astype(float)
display(pd.DataFrame([{"dataset": ds, "n_features": len(c), "features": ", ".join(c)} for ds, c in FEATURES.items()]))
````

### F6. Shared analysis functions (used identically for NPS in Part G and CSAT in Part H)

````python
def survey_frame(metric):
    """Observed responses of one metric with score, category and the bad/good flags (no imputation: rows without the metric are excluded)."""
    sp = SURVEY_SPECS[metric]
    S = RESP[RESP[sp["col"]].notna()].copy()
    S["score"] = S[sp["col"]].astype(float)
    S["cat"] = pd.cut(S.score, sp["bins"], labels=sp["cats"]).astype(str)
    S["is_bad"], S["is_good"] = (S.cat == sp["bad"]).astype(int), (S.cat == sp["good"]).astype(int)
    return S, sp

def net_index(s_cat, sp):
    n = len(s_cat)
    return round(100.0 * ((s_cat == sp["good"]).sum() - (s_cat == sp["bad"]).sum()) / n, 1) if n >= MIN_N_INDEX else np.nan

def score_summary(g, sp):
    return pd.Series({"n": len(g), "avg_score": g.score.mean() if len(g) >= MIN_N_GROUP else np.nan,
                      f"{sp['bad']}_pct": 100 * g.is_bad.mean() if len(g) >= MIN_N_GROUP else np.nan,
                      f"{sp['good']}_pct": 100 * g.is_good.mean() if len(g) >= MIN_N_GROUP else np.nan,
                      sp["index_name"]: net_index(g.cat, sp), "suppressed": len(g) < MIN_N_INDEX})

def profile_by_category(S, sp, feats):
    """Doc §4.1/§4.2 tables: mean (and median) of each behavioural feature per survey category."""
    agg = S.groupby("cat")[feats].agg(["mean", "median"])
    agg.columns = [f"{c}_{s}" for c, s in agg.columns]
    out = S.groupby("cat").size().rename("n_responses").to_frame().join(agg).reindex(list(sp["cats"]))
    return out.reset_index().rename(columns={"cat": f"{sp['label'].lower()}_category"})

def feature_association(S, sp, feats):
    """Per feature: Spearman with the raw score, Kruskal-Wallis effect size across categories, bad-vs-good means and Cohen's d."""
    rows = []
    for c in feats:
        d = S[[c, "score", "cat"]].dropna()
        if len(d) < MIN_N_GROUP or d[c].nunique() < 2:
            continue
        rho, p = spearmanr(d[c], d.score)
        groups = [g[c].values for _, g in d.groupby("cat") if len(g) >= MIN_N_GROUP]
        H_, pk = kruskal(*groups) if len(groups) >= 2 else (np.nan, np.nan)
        eps2 = (H_ - len(groups) + 1) / (len(d) - len(groups)) if len(groups) >= 2 else np.nan
        b, g_ = d.loc[d.cat == sp["bad"], c], d.loc[d.cat == sp["good"], c]
        sd = np.sqrt(((len(b) - 1) * b.var() + (len(g_) - 1) * g_.var()) / max(len(b) + len(g_) - 2, 1))
        rows.append({"dataset": FEATURE_DATASET.get(c, ""), "feature": c, "n": len(d), "coverage_pct": round(100 * len(d) / len(S), 2),
                     "spearman_rho": rho, "p_value": p, "kruskal_epsilon2": eps2,
                     f"mean_{sp['bad']}": b.mean(), f"mean_{sp['good']}": g_.mean(),
                     f"cohens_d_{sp['good']}_vs_{sp['bad']}": (g_.mean() - b.mean()) / sd if sd > 0 else np.nan})
    out = pd.DataFrame(rows)
    if len(out):
        out["strength"] = pd.cut(out.spearman_rho.abs(), [-1, .1, .3, .5, 1.01], labels=["negligible", "weak", "moderate", "strong"])
    return out

BUCKETS = {   # feature -> (bins, labels); right-closed bins.  None = categorical column used as-is
    "n_tx":                  ([-0.5, 0.5, 1.5, 3.5, 6.5, np.inf], ["0", "1", "2-3", "4-6", "7+"]),
    "n_pay_failed":          ([-0.5, 0.5, 1.5, np.inf], ["0", "1", "2+"]),
    "n_refund":              ([-0.5, 0.5, 1.5, np.inf], ["0", "1", "2+"]),
    "gross_spend_pctile":    ([-0.1, 20, 40, 60, 80, 100], ["Q1 (lowest)", "Q2", "Q3", "Q4", "Q5 (highest)"]),
    "avg_tx_value_pctile":   ([-0.1, 20, 40, 60, 80, 100], ["Q1 (lowest)", "Q2", "Q3", "Q4", "Q5 (highest)"]),
    "days_since_last_tx":    ([-0.5, 7, 30, 90, np.inf], ["0-7d", "8-30d", "31-90d", "91d+"]),
    "tx_trend":              None,
    "n_eng":                 ([-0.5, 0.5, 2.5, 5.5, 10.5, np.inf], ["0", "1-2", "3-5", "6-10", "11+"]),
    "n_login":               ([-0.5, 0.5, 1.5, 4.5, np.inf], ["0", "1", "2-4", "5+"]),
    "n_app_events":          ([-0.5, 0.5, 2.5, np.inf], ["0", "1-2", "3+"]),
    "n_web_events":          ([-0.5, 0.5, 2.5, np.inf], ["0", "1-2", "3+"]),
    "avg_session_sec":       ([0, 60, 180, 300, np.inf], ["<1m", "1-3m", "3-5m", "5m+"]),
    "n_optout":              ([-0.5, 0.5, np.inf], ["0", "1+"]),
    "days_since_last_eng":   ([-0.5, 7, 30, 90, np.inf], ["0-7d", "8-30d", "31-90d", "91d+"]),
    "eng_trend":             None,
    "repeat_contact":        ([-0.5, 0.5, 1.5], ["first_contact", "repeat_contact"]),
    "source_combo":          None,   # doc §1 table: direct relationship with source_combo / n_sources / active
    "n_sources":             None,
}
NULL_LABEL = {"gross_spend_pctile": "no_successful_payment", "avg_tx_value_pctile": "no_successful_payment",
              "days_since_last_tx": "no_prior_tx_in_window", "days_since_last_eng": "no_prior_eng_in_window",
              "avg_session_sec": "no_session"}

TREND_ORDER = ["declining", "stable", "increasing", "new_activity", "no_prior_activity", "insufficient_history"]

def bucketise(S, feat):
    """Ordered categorical bucket; NULL gets an explicit label (never merged into a numeric bucket)."""
    null_lbl = NULL_LABEL.get(feat, "not_available")
    spec = BUCKETS[feat]
    if spec is None:
        b, order = S[feat].astype(object).where(S[feat].notna(), null_lbl), TREND_ORDER
    else:
        b, order = pd.cut(S[feat], spec[0], labels=spec[1]).astype(object).where(S[feat].notna(), null_lbl), list(spec[1])
    cats = [c for c in order if c in set(b)] + sorted(set(b) - set(order) - {null_lbl}) + ([null_lbl] if null_lbl in set(b) else [])
    return pd.Categorical(b, categories=cats, ordered=True)

def score_by_buckets(S, sp, feats):
    """Doc §1.1/§2.1: average score, bad/good share and net index per behavioural bucket."""
    out = []
    for f in feats:
        if f not in S.columns or S[f].isna().all():
            continue
        g = S.assign(bucket=bucketise(S, f)).groupby("bucket", observed=True).apply(score_summary, sp=sp, include_groups=False)
        ds = FEATURE_DATASET.get(f, {"tx_trend": "transactions", "eng_trend": "engagements"}.get(f, "cross-dataset"))
        out.append(g.reset_index().assign(dataset=ds, feature=f))
    cols = ["dataset", "feature", "bucket", "n", "avg_score", f"{sp['bad']}_pct", f"{sp['good']}_pct", sp["index_name"], "suppressed"]
    return pd.concat(out, ignore_index=True)[cols] if out else pd.DataFrame(columns=cols)

def high_vs_low(S, sp, feat, q_=0.25):
    """Top vs bottom quartile among responses with the activity present (doc §1.1 high- vs low-frequency transactors)."""
    d = S[S[feat] > 0]
    if len(d) < 4 * MIN_N_GROUP:
        return pd.DataFrame()
    lo, hi = d[feat].quantile(q_), d[feat].quantile(1 - q_)
    grp = np.where(d[feat] <= lo, f"low (<= {lo:g})", np.where(d[feat] >= hi, f"high (>= {hi:g})", "middle"))
    return d.assign(group=grp).groupby("group").apply(score_summary, sp=sp, include_groups=False).reset_index().assign(feature=feat)

def event_type_presence(S, sp):
    """Doc §2.1 'by engagement type': score when an event type occurred in the survey month vs not (among engaged months)."""
    d, rows = S[S.n_eng > 0], []
    for e, c in EVT_COLS:
        if c not in d.columns:
            continue
        w, wo = d[d[c] > 0], d[d[c] == 0]
        if len(w) >= MIN_N_GROUP and len(wo) >= MIN_N_GROUP:
            rows.append({"event_type": e, "n_with": len(w), "avg_score_with": w.score.mean(), "avg_score_without": wo.score.mean(),
                         "diff": w.score.mean() - wo.score.mean(), f"{sp['index_name']}_with": net_index(w.cat, sp),
                         f"{sp['index_name']}_without": net_index(wo.cat, sp)})
    return pd.DataFrame(rows).sort_values("diff") if rows else pd.DataFrame()

OUTCOMES = {"y_no_tx_next": "no transaction", "y_no_eng_next": "no engagement", "y_no_int_next": "no interaction",
            "y_inactive_next": "fully inactive (A9 definition)", "y_tx_decline_next": "transactions decline >30%",
            "y_eng_decline_next": "engagement declines >30%"}

def future_outcomes(S, sp, by=("cat",)):
    """Doc §1.2/§2.2/§3.1: separate future outcomes per survey category (rates in %, n per outcome = full-horizon rows)."""
    d = S[S.has_full_horizon]
    agg = {f"pct_{k}": (k, lambda s: 100 * s.mean()) for k in OUTCOMES}
    agg.update({"n": ("score", "size"), "avg_fut_n_tx": ("fut_n_tx", "mean"), "avg_fut_n_eng": ("fut_n_eng", "mean"),
                "avg_n_tx_prior": ("n_tx_prior", "mean"), "avg_n_eng_prior": ("n_eng_prior", "mean")})
    out = d.groupby(list(by)).agg(**agg).reset_index()
    return out[out.n >= MIN_N_GROUP]

def outcome_lift(fo, sp):
    """Bad-vs-good ratio for each separated outcome (how strongly the survey separates each dataset's future behaviour)."""
    f = fo.set_index("cat")
    if sp["bad"] not in f.index or sp["good"] not in f.index:
        return pd.DataFrame()
    return pd.DataFrame([{"outcome": k, "description": v, f"rate_{sp['bad']}_pct": f.loc[sp["bad"], f"pct_{k}"],
                          f"rate_{sp['good']}_pct": f.loc[sp["good"], f"pct_{k}"],
                          "lift_bad_vs_good": f.loc[sp["bad"], f"pct_{k}"] / f.loc[sp["good"], f"pct_{k}"]
                          if f.loc[sp["good"], f"pct_{k}"] > 0 else np.nan} for k, v in OUTCOMES.items()])

ASSOC = {}                        # metric -> dataset -> feature_association table (filled in G/H, compared in I2)
BEHAVIOUR_MODEL_FEATURES = None   # set below once (same list for both metrics and the comparison)
def model_auc(df, feats, target):
    """Customer-grouped hold-out AUC (train/test from split_c). HistGradientBoosting handles NULL natively -> no imputation."""
    d = df[df[target].notna()]
    tr, te = d[d.split == "train"], d[d.split == "test"]
    if tr[target].nunique() < 2 or te[target].nunique() < 2 or len(te) < MIN_N_GROUP:
        return np.nan
    m = HistGradientBoostingClassifier(max_iter=200, learning_rate=0.1, random_state=RANDOM_STATE)
    m.fit(tr[feats], tr[target].astype(int))
    return roc_auc_score(te[target].astype(int), m.predict_proba(te[feats])[:, 1])

def incremental_value(S, sp, extra=()):
    """Doc §4.3: does the survey add information beyond text sentiment AND behavioural features? (future outcomes, test customers)."""
    beh = BEHAVIOUR_MODEL_FEATURES
    sets = {"behaviour": beh, "behaviour+text": beh + ["text_pred"], f"behaviour+text+{sp['label']}": beh + ["text_pred", "score"]}
    for name, cols in extra:
        sets[name] = cols
    rows = []
    for tgt in ("y_inactive_next", "y_no_tx_next", "y_no_eng_next", "y_tx_decline_next"):
        r = {"target": tgt, "n_test": int(((S.split == "test") & S[tgt].notna()).sum()), "base_rate_pct": 100 * S[tgt].mean()}
        for name, cols in sets.items():
            r[f"auc_{name}"] = model_auc(S, cols, tgt)
        r[f"delta_auc_{sp['label']}_vs_behaviour+text"] = r[f"auc_behaviour+text+{sp['label']}"] - r["auc_behaviour+text"]
        rows.append(r)
    return pd.DataFrame(rows)

def positive_but_declining(S, sp):
    """Doc §4.3: good survey category but declining trailing behaviour -> should be flagged? Compared on future outcomes."""
    d = S[S.has_full_horizon].copy()
    d["declining_behaviour"] = (d.tx_trend == "declining") | (d.eng_trend == "declining")
    d["group"] = np.select([d.is_good.astype(bool) & d.declining_behaviour, d.is_good.astype(bool),
                            d.is_bad.astype(bool) & d.declining_behaviour, d.is_bad.astype(bool)],
                           [f"{sp['good']} + declining", f"{sp['good']} + not declining",
                            f"{sp['bad']} + declining", f"{sp['bad']} + not declining"], default="other")
    agg = {f"pct_{k}": (k, lambda s: 100 * s.mean()) for k in ("y_inactive_next", "y_no_tx_next", "y_no_eng_next", "y_tx_decline_next")}
    out = d[d.group != "other"].groupby("group").agg(n=("score", "size"), **agg).reset_index()
    return out[out.n >= MIN_N_GROUP]

def text_then_purchasing(S, sp):
    """Doc §4.3: is negative interaction sentiment followed by declining purchasing? (within each survey category)."""
    d = S[S.has_full_horizon & (S.n_prior_months == OUTCOME_HORIZON)]
    out = d.groupby(["text_label_pred", "cat"]).agg(n=("score", "size"),
                                                    pct_tx_decline_next=("y_tx_decline_next", lambda s: 100 * s.mean()),
                                                    avg_fut_minus_prior_tx=("fut_n_tx", "mean"), avg_prior_tx=("n_tx_prior", "mean")).reset_index()
    out["avg_fut_minus_prior_tx"] = out.avg_fut_minus_prior_tx - out.avg_prior_tx
    return out[out.n >= MIN_N_GROUP].drop(columns="avg_prior_tx")

def term_lift(S, sp, top=15):
    """Doc §3.1 'issue-level NLP features' (exploratory, TRAIN customers only): n-grams over-represented in bad vs good responses."""
    from sklearn.feature_extraction.text import CountVectorizer
    d = S[(S.split == "train") & S.cat.isin([sp["bad"], sp["good"]])]
    t = q("SELECT interaction_id, txt FROM itx").set_index("interaction_id").txt
    d = d.assign(txt=d.interaction_id.map(t)).dropna(subset=["txt"])
    g = d.groupby(["txt", "cat"]).size().unstack(fill_value=0).reindex(columns=[sp["bad"], sp["good"]], fill_value=0)
    if len(g) < 2:
        return pd.DataFrame()
    cv = CountVectorizer(ngram_range=(1, 2), min_df=1, binary=True)
    X = cv.fit_transform(g.index)
    nb, ng = X.T @ g[sp["bad"]].values, X.T @ g[sp["good"]].values
    lor = np.log((nb + 0.5) / (g[sp["bad"]].sum() + 0.5)) - np.log((ng + 0.5) / (g[sp["good"]].sum() + 0.5))
    tl = pd.DataFrame({"term": cv.get_feature_names_out(), f"n_{sp['bad']}": nb, f"n_{sp['good']}": ng, "log_odds_bad_vs_good": lor})
    tl = tl[(tl[f"n_{sp['bad']}"] + tl[f"n_{sp['good']}"]) >= MIN_N_GROUP]
    return pd.concat([tl.nlargest(top, "log_odds_bad_vs_good").assign(direction=f"over-represented in {sp['bad']}"),
                      tl.nsmallest(top, "log_odds_bad_vs_good").assign(direction=f"over-represented in {sp['good']}")])

def by_dimension_month(S, sp, dims=("loyalty_tier", "customer_segment", "region")):
    """Doc §3.1 'monitor fairness / continue reporting': net index per profile level and month (suppressed below MIN_N_INDEX)."""
    out = []
    for dm in dims:
        g = S.groupby([dm, "mth"]).apply(score_summary, sp=sp, include_groups=False).reset_index().rename(columns={dm: "level"})
        out.append(g.assign(dimension=dm))
    return pd.concat(out, ignore_index=True)

def regional_drivers(S, sp):
    """Doc §3.1 'investigate regional drivers': region x interaction type net index + region x key behaviour means."""
    a = S.groupby(["region", "interaction_type"]).apply(score_summary, sp=sp, include_groups=False).reset_index()
    a["gap_vs_type_overall"] = a.avg_score - a.interaction_type.map(S.groupby("interaction_type").score.mean())
    return a

def resolution_time(S, sp):
    """Doc §3.1 'examine resolution time' - only when an interaction duration column exists (F1)."""
    if "resolution_time_sec" not in S.columns or S.resolution_time_sec.isna().all():
        return pd.DataFrame()
    b = pd.qcut(S.resolution_time_sec.rank(method="first"), 5, labels=["Q1 (fastest)", "Q2", "Q3", "Q4", "Q5 (slowest)"])
    return S.assign(bucket=b).groupby(["resolution_status", "bucket"], observed=True).apply(
        score_summary, sp=sp, include_groups=False).reset_index()

def qa_answer(assoc, feat):
    """Plain-English answer to 'do the bad-category respondents have LOWER <feat>?' from Cohen's d (good - bad)."""
    r = assoc[assoc.feature == feat] if len(assoc) else assoc
    if not len(r):
        return "n/a (feature not available)"
    d = r.filter(like="cohens_d").iloc[0, 0]
    size = "negligible" if abs(d) < 0.1 else "small" if abs(d) < 0.3 else "medium" if abs(d) < 0.5 else "large"
    verdict = "no material difference" if size == "negligible" else ("yes" if d > 0 else "no (opposite direction)")
    return f"{verdict} - {size} effect (d={d:+.2f})"

def plot_profile(S, sp, feats, fname):
    """Index (category mean / overall mean x 100) for the key features of each dataset."""
    feats = [f for f in feats if f in S.columns and S[f].notna().any() and S[f].mean() not in (0, np.nan)]
    if not feats:
        return
    m = S.groupby("cat")[feats].mean().reindex(list(sp["cats"]))
    idx = (m / S[feats].mean() * 100).T
    colors = {sp["cats"][0]: "#C00000", sp["cats"][1]: "#BFBFBF", sp["cats"][2]: "#00B050"}
    ax = idx.plot.barh(figsize=(11, max(4, 0.45 * len(feats))), color=[colors[c] for c in idx.columns])
    ax.legend(title=f"{sp['label']} category")
    ax.axvline(100, color="black", lw=.8); ax.set_xlabel("index vs all respondents (100 = average)")
    ax.set_title(f"{sp['label']}: behavioural profile by category (survey month)")
    plt.tight_layout(); plt.savefig(OUT_DIR / f"{fname}.png", dpi=120); plt.show()

def plot_outcomes(fo, sp, fname):
    cols = [f"pct_{k}" for k in ("y_no_tx_next", "y_no_eng_next", "y_no_int_next", "y_inactive_next")]
    f = fo.set_index("cat").reindex(list(sp["cats"]))[cols]
    f.columns = [OUTCOMES[c[4:]] for c in cols]
    ax = f.T.plot.bar(figsize=(11, 4), color=["#C00000", "#BFBFBF", "#00B050"], rot=0)
    ax.legend(title=f"{sp['label']} category")
    ax.set_ylabel(f"% in next {OUTCOME_HORIZON} months"); ax.set_title(f"{sp['label']}: separated future outcomes by category")
    ax.grid(alpha=.3, axis="y"); plt.tight_layout(); plt.savefig(OUT_DIR / f"{fname}.png", dpi=120); plt.show()

KEY_TX = [c for c in ("n_tx", "n_pay_success", "n_pay_failed", "n_refund", "gross_spend_pctile", "avg_tx_value_pctile", "days_since_last_tx")
          if c in FEATURE_DATASET]
KEY_ENG = [c for c in ("n_eng", "n_login", "n_app_events", "n_web_events", "n_email_click", "n_campaign_responses", "avg_session_sec",
                       "n_optout", "days_since_last_eng") if c in FEATURE_DATASET]
# behaviour-only model inputs: transaction + engagement + interaction features except text (text is added as its own step)
BEHAVIOUR_MODEL_FEATURES = [c for c in FEATURES["transactions"] + FEATURES["engagements"] + FEATURES["interactions"] if c != "text_pred"]
print("behaviour model features:", len(BEHAVIOUR_MODEL_FEATURES))
````

# Part I – Comparative analysis: NPS vs CSAT across features and datasets
Parts G and H answer the document's questions for each metric on its **own** respondents. Because NPS (≈12 % of interactions) and
CSAT (≈45 %) are collected independently (A1), their respondent sets differ. Every comparison below is therefore shown
(a) on each metric's own responses and, where it matters, (b) on the **both-observed** subset (same interactions) – the only
like-for-like comparison. Levels are compared on the model's −1…+1 scale (`csat_norm`, `nps_norm`) **and** as a scale-free z-score
versus each metric's own mean (the two scales have different neutral points, so raw normalised gaps are partly a scale artefact).

| Step | Comparison | Datasets |
|---|---|---|
| I1 | Survey coverage by behavioural source (who can be scored by NPS vs CSAT) | customers × transactions × engagements × interactions |
| I2 | Feature association side by side (rho, effect size, rank agreement) incl. both-observed subset | transactions, engagements, interactions |
| I3 | Separated future outcome lift (bad vs good category) | transactions, engagements, interactions |
| I4 | Incremental predictive value: +CSAT, +NPS, +both beyond behaviour + text (both-observed subset) | all four |
| I5 | Profile dimensions (tier, segment, region): net index and level ranking | customers |
| I6 | Interaction dimensions (type, resolution, FCR, channel, repeat contact, text label) | interactions |
| I7 | Disagreement cases (CSAT satisfied but NPS detractor, and the reverse) | interactions + future behaviour |
| I8 | Summary chart and comparative findings | – |

### I1. Survey coverage by behavioural source (customer-month level, active months)

````python
i1 = []
for dim in ("source_combo", "has_tx", "has_eng", "has_int", "loyalty_tier"):
    t = q(f"""SELECT '{dim}' dimension, CAST({dim} AS VARCHAR) level, count(*) active_cm,
                     round(100.0*count_if(has_nps)/count(*), 2) pct_with_nps, round(100.0*count_if(has_csat)/count(*), 2) pct_with_csat,
                     round(100.0*count_if(has_nps AND has_csat)/count(*), 2) pct_with_both,
                     round(100.0*count_if(has_nps OR has_csat)/count(*), 2) pct_with_any
              FROM cm_voice WHERE active GROUP BY ALL ORDER BY active_cm DESC""")
    i1.append(t)
i1 = save(pd.concat(i1, ignore_index=True), "I1_survey_coverage_by_source", show=20)
no_int = i1[(i1.dimension == "has_int") & (i1.level.str.lower() == "false")]
if len(no_int):
    flag("I1", f"Active months without an interaction: NPS {no_int.pct_with_nps.iloc[0]}% / CSAT {no_int.pct_with_csat.iloc[0]}% "
               "-> both surveys only exist where there is an interaction; TX/ENG-only months must be scored on behaviour")
````

### I2. Feature association – NPS vs CSAT side by side (own respondents and both-observed subset)

````python
S_both = RESP[RESP.nps.notna() & RESP.csat.notna()].copy()
print(f"both-observed responses: {len(S_both):,}")
rows = []
for ds, feats in FEATURES.items():
    for f in feats:
        r = {"dataset": ds, "feature": f}
        for m in ("nps", "csat"):
            a = ASSOC.get(m, {}).get(ds, pd.DataFrame())
            a = a[a.feature == f] if len(a) else a
            r[f"rho_{m}_own"] = a.spearman_rho.iloc[0] if len(a) else np.nan
            r[f"eps2_{m}_own"] = a.kruskal_epsilon2.iloc[0] if len(a) else np.nan
            r[f"d_{m}_own"] = a.filter(like="cohens_d").iloc[0, 0] if len(a) else np.nan
            d = S_both[[f, m]].dropna()
            r[f"rho_{m}_both"] = spearmanr(d[f], d[m])[0] if len(d) >= MIN_N_GROUP and d[f].nunique() > 1 else np.nan
        r["n_both"] = int(S_both[f].notna().sum())
        r["abs_rho_diff_both"] = abs(r["rho_nps_both"]) - abs(r["rho_csat_both"])
        r["stronger_link_both"] = ("NPS" if r["abs_rho_diff_both"] > 0.02 else "CSAT" if r["abs_rho_diff_both"] < -0.02 else "similar")
        rows.append(r)
CMP = save(pd.DataFrame(rows), "I2_association_nps_vs_csat", show=40)
rank_rho = spearmanr(CMP.rho_nps_own.abs(), CMP.rho_csat_own.abs(), nan_policy="omit")[0] if len(CMP) > 2 else np.nan
by_ds = save(CMP.groupby("dataset").agg(n_features=("feature", "size"),
                                         mean_abs_rho_nps_own=("rho_nps_own", lambda s: s.abs().mean()),
                                         mean_abs_rho_csat_own=("rho_csat_own", lambda s: s.abs().mean()),
                                         mean_abs_rho_nps_both=("rho_nps_both", lambda s: s.abs().mean()),
                                         mean_abs_rho_csat_both=("rho_csat_both", lambda s: s.abs().mean()),
                                         features_nps_stronger=("stronger_link_both", lambda s: (s == "NPS").sum()),
                                         features_csat_stronger=("stronger_link_both", lambda s: (s == "CSAT").sum())).reset_index(),
             "I2_association_by_dataset")
flag("I2", f"Feature-importance ranking agreement NPS vs CSAT (Spearman of |rho|): {rank_rho:.2f}; mean |rho| by dataset (both-observed) - "
           + "; ".join(f"{r.dataset}: NPS {r.mean_abs_rho_nps_both:.3f} vs CSAT {r.mean_abs_rho_csat_both:.3f}" for r in by_ds.itertuples()))
````

### I4. Incremental predictive value on the same responses: +CSAT, +NPS, +both beyond behaviour + text
Both-observed subset only (so neither survey is ever imputed); customer-grouped hold-out; HistGradientBoosting (native NULL handling).

````python
Sb = S_both.copy()
base = BEHAVIOUR_MODEL_FEATURES + ["text_pred"]
i4 = []
for tgt in ("y_inactive_next", "y_no_tx_next", "y_no_eng_next", "y_tx_decline_next"):
    r = {"target": tgt, "n_test": int(((Sb.split == "test") & Sb[tgt].notna()).sum())}
    for name, cols in (("behaviour+text", base), ("+CSAT", base + ["csat"]), ("+NPS", base + ["nps"]),
                       ("+CSAT+NPS", base + ["csat", "nps"]), ("CSAT only", ["csat"]), ("NPS only", ["nps"])):
        r[f"auc_{name}"] = model_auc(Sb, cols, tgt)
    r["delta_csat"] = r["auc_+CSAT"] - r["auc_behaviour+text"]
    r["delta_nps"] = r["auc_+NPS"] - r["auc_behaviour+text"]
    r["delta_nps_given_csat"] = r["auc_+CSAT+NPS"] - r["auc_+CSAT"]
    r["delta_csat_given_nps"] = r["auc_+CSAT+NPS"] - r["auc_+NPS"]
    i4.append(r)
i4 = save(pd.DataFrame(i4), "I4_incremental_value_nps_vs_csat")
r0 = i4.set_index("target").loc["y_inactive_next"]
flag("I4", f"Fully-inactive-next AUC (both-observed, test): behaviour+text {r0['auc_behaviour+text']:.4f}, +CSAT {r0['delta_csat']:+.4f}, "
           f"+NPS {r0['delta_nps']:+.4f}, NPS given CSAT {r0['delta_nps_given_csat']:+.4f}, CSAT given NPS {r0['delta_csat_given_nps']:+.4f}")
````

### I5. Profile dimensions (customers dataset): net index and level ranking, NPS vs CSAT

````python
def dim_compare(dims, frame_fn=survey_frame):
    out = []
    for dm in dims:
        parts = []
        for m in ("nps", "csat"):
            S_, sp_ = frame_fn(m)
            mu, sd = S_.score.mean(), S_.score.std()
            g = S_.groupby(dm).apply(lambda x: pd.Series({"n": len(x), "avg_norm": x[sp_["norm"]].mean(),
                                                         "z_vs_own_mean": (x.score.mean() - mu) / sd,   # scale-free position
                                                         "net_index": net_index(x.cat, sp_),
                                                         "bad_pct": 100 * x.is_bad.mean()}), include_groups=False)
            parts.append(g.add_prefix(f"{m}_"))
        t = pd.concat(parts, axis=1).reset_index().rename(columns={dm: "level"}).assign(dimension=dm)
        t["gap_z_nps_minus_csat"] = t.nps_z_vs_own_mean - t.csat_z_vs_own_mean   # >0: level looks better on NPS than on CSAT
        t["rank_nps"], t["rank_csat"] = t.nps_avg_norm.rank(ascending=False), t.csat_avg_norm.rank(ascending=False)
        out.append(t)
    res = pd.concat(out, ignore_index=True)
    agree = res.groupby("dimension").apply(lambda d: spearmanr(d.nps_avg_norm, d.csat_avg_norm)[0] if len(d) > 2 else np.nan,
                                           include_groups=False).rename("rank_agreement_rho").reset_index()
    return res, agree

i5, i5a = dim_compare(("loyalty_tier", "customer_segment", "region"))
save(i5, "I5_profile_dimensions_nps_vs_csat", show=20)
save(i5a, "I5_profile_rank_agreement")
````
